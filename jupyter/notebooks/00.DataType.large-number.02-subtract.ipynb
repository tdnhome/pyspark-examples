- You are given two exceedingly large positive decimal numbers, num1 and num2, both represented as strings.
- The length of these strings can range anywhere from 1 to 500 characters.
- The challenge here is to subtract num2 from num1 without directly converting the strings into integers.

***Task***

- Create a Python function that performs this operation and returns the resultant string, referred to as num3.

***Hint***

- The subtraction will not result in a negative number, as num1 will always be greater than or equal to num2.

### Solution Building

```python
# Solution Building: Step 1
# -----------------------------------
# Before we start coding, let's consider the strategy we're going to adopt. 
# You may recall that each digit in a number has a value, and the position of the digit determines 
# its influence on the total value of the number. This system is called place-value notation.

# The first step requires initialization of our variables. We'll use two indices, i and j, 
# to point to the current digit in num1 and num2, respectively. 
# We'll also need a variable borrow to hold borrow from each subtraction operation. 
# Lastly, we'll employ a list named res to hold our result, where each digit from the addition is appended at the front.

def solution(num1, num2):
    # We will scan numbers from the least significant digit to the most significant one (right to left)
    i = len(num1) - 1 # We will scan through num1 and num2 from right to left
    j = len(num2) - 1 # We will scan through num1 and num2 from right to left
    borrow = 0
    res = [] # the resulting list of integers (as strings) in with the order of significant digits reversed


# Solution Building: Step 2
# -----------------------------------
# Scanning through num1 and num2 from right to left, from the least significant digit to the most significant one.
# For each iteration, we extract the digits n1 from num1 and n2 from num2. 
# If i or j is less than 0, we've processed all the digits in one of the numbers.
# Consequently, we consider these additional digits as 0.

    # Note that these are OR conditions
    while i >= 0 or j >= 0:
        n1 = int(num1[i]) if i >= 0 else 0  # set to 0 if there is no number in the current position
        n2 = int(num2[j]) if j >= 0 else 0  # set to 0 if there is no number in the current position

# Solution Building: Step 3
# -----------------------------------
# After obtaining the digits n1 and n2, our immediate step is to subract them. However, the borrow, 
# which accumulates any underflow from the subtraction of previous column digits, must also be considered. 
# Subsequently, we append curr to res and decrement both i and j before embarking on the next iteration. 
# Finally, we reverse res and join the digits into a single string to acquire our result.

        if borrow > 0:    # from previous loop
            n1 -= 1
            borrow = 0

        if n1 < n2:       # determine if we need to borrow
            n1 += 10
            borrow = 1    # for next loop
        
        diff = n1 - n2

        res.append(str(diff))
        
        i -= 1
        j -= 1

    # res is the list integers (as strings) in with the order of significant digits reversed
    result = ''.join(res[::-1]) # reverse the list and join into a single string
    result.lstrip('0') or '0'   # remove leading zeroes from the string
    print(f"Result: {result}")
    return result.lstrip('0') or '0'


# Tests
assert solution('9', '1') == '8'
assert solution('100', '1') == '99'
assert solution('501', '105') == '396'
assert solution('1000', '999') == '1'
assert solution('10000', '1000') == '9000'
assert solution('111111111111111', '111111111111111') == '0'
assert solution('398746', '199234') == '199512'
assert solution('123456789101112131415', '98765432101012131415') == '24691357000100000000'
assert solution('500', '500') == '0'
```

## Introduction

***Introduction***

You will be given two arrays of integers. 
- The first array has n elements, range from 1 to 100, inclusive
- The second array has k elements, range from 1 to 100, inclusive
- The elements of both arrays can fall within a range of -100 to 100, inclusive.

***Task***

Your task is to write a Python function that will locate and return an array of all pairs of integers with the property that 
- The first element of each pair comes from the first array and
- The second element of each pair comes from the second array
- The sum of the two elements of the pair is a perfect square.
  - A perfect square is an integer that is the square of another integer.
- The order of pairs in your output should correspond to the order of the elements in the input arrays.
- If no such pairs exist, or if either input array is empty, your function should return an empty list.

***Example 1***

- If the two arrays are [2, 3, 16] and [1, 9, 10],
- then the function should return [(3, 1), (16, 9)]
  - because 3+1=4 (which is the square of 2) and 16+9=25 (which is the square of 5).

***Example 2***

If you have array1 = [0, 5] and arrat2 = [4, 11], the pairs would be:
- (0, 4) because 0+4=4 (which is 2 to the power of 2)
- (5, 4) because 5+4=9 (which is 3 to the power of 2)
- (5, 11) because 5+11=16 (which is 4 to the power of 2) 

The result would be [(0, 4), (5, 4), (5, 11)].

## Hint: check if a number is a perfect square

```python
import math

def is_perfect_square(n):
    if n < 0:
        return False  # Negative numbers cannot be perfect squares
    
    # Get the exact integer square root
    root = math.isqrt(n)
    
    # Check if squaring the root yields the original number
    return root * root == n
    
print(is_perfect_square(16))          # True
print(is_perfect_square(20))          # False
print(is_perfect_square(0))           # True
```

## Solution

```python
def solution(arrayA, arrayB):

    result = []
    for itemA in arrayA:
        for itemB in arrayB:
            if (is_perfect_square(itemA + itemB)):
                pair = (itemA, itemB)
                if pair not in result:
                    result.append(pair)
    
    print(f"result: {result}")
    return result


import math

def is_perfect_square(n):
    if n < 0:
        return False  # Negative numbers cannot be perfect squares
    
    # Get the exact integer square root
    root = math.isqrt(n)
    
    # Check if squaring the root yields the original number
    return root * root == n


# Test from the introduction secion
arr1 = [2, 3, 16] 
arr2 = [1, 9, 10]
expected_output =  [(3, 1), (16, 9)]
assert solution(arr1, arr2) == expected_output

arr1 = [0, 5]
arr2 = [4, 11]
expected_output = [(0, 4), (5, 4), (5, 11)]
assert solution(arr1, arr2) == expected_output

# Tests from https://codesignal.com/learn/course/93/unit/1/practice/4
arr1 = [2, 3, 16]
arr2 = [1, 9, 10]
expected_output = [(3, 1), (16, 9)]
assert solution(arr1, arr2) == expected_output

arr1 = [0]
arr2 = [0]
expected_output = [(0, 0)]
assert solution(arr1, arr2) == expected_output

arr1 = [4, 13, 23]
arr2 = [-4, -3, -24]
expected_output = [(4, -4), (4, -3), (13, -4)]
assert solution(arr1, arr2) == expected_output

arr1 = [0, 1, 2, -100, 100]
arr2 = [-100, 100, 30, 0, -1, -2, -3]
expected_output = [(0, 100), (0, 0), (1, 0), (1, -1), (2, -1), (2, -2), (-100, 100), (100, -100), (100, 0)]
assert solution(arr1, arr2) == expected_output

arr1 = [100, 75, 36, 9, -25, -64, -100]
arr2 = [-1, 1, 24, 0, -1, -24]
expected_output = [(100, 0), (36, 0), (9, 0)]
assert solution(arr1, arr2) == expected_output

arr1 = []
arr2 = [1, 2, 3, 4]
expected_output = []
assert solution(arr1, arr2) == expected_output

arr1 = [1, 2, 3, 4]
arr2 = []
expected_output = []
assert solution(arr1, arr2) == expected_output

arr1 = []
arr2 = []
expected_output = []
assert solution(arr1, arr2) == expected_output
```

## Concurrency vs Parallelism

***Concurrency***: multiple tasks make progress during overlapping time periods.

Example:
- Web server handling thousands of requests
- Chat application
- File downloads

Tasks may switch execution frequently.

***Parallelism***: tasks literally run simultaneously on multiple CPU cores.

Example:
- Image processing
- AI/ML computation


## Python GIL (Global Interpreter Lock)

One of the most important concepts in Python concurrency.

***What is GIL?***" only one thread executes Python bytecode at a time.

This means:
- Python threads are NOT truly parallel for CPU-heavy work
- Threads are still useful for I/O tasks

***Why GIL Exists***

The GIL simplifies:
- Memory management
- Reference counting
- Thread safety inside the interpreter

***Impact of GIL***
|Task Type	|Best Choice|
|-----------|-----------|
|I/O-bound	|Threading (Multi threading), Asyncio (Async IO), aiohttp (Async HTTP)|
|CPU-bound	|Multi processing (true parallelism using multiple processes (CPU cores)|

## Threading vs Multiprocessing

from concurrent.futures import:
- ThreadPoolExecutor (multi-threading)
- ProcessPoolExecutor (multi-processing)

|Feature	|Threading	|Multiprocessing|
|-----------|-----------|---------------|
|Memory	|Shared	|Separate|
|GIL	|Affected	|Not affected|
|Speed	|Good for I/O	|Good for CPU|
|Communication	|Easy	|More complex|
|Startup cost	|Low	|Higher|

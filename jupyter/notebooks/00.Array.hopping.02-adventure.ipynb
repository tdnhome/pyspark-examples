## Problem Statement

***Introduction***

Gloria the bunny finds herself once again amidst an array game. This time, however, the game has slightly intensified with a third array coming into play. Your task is to develop a Python function to maneuver Gloria through her quest, yielding the summation of the maximum values she encounters from arrayB and arrayC together.

Gloria's journey begins at the first element of arrayA. Gloria's movement pattern follows a fixed sequence that repeats: arrayA -> arrayB -> arrayA -> arrayC. In other words, Gloria always alternates between arrayA and either arrayB or arrayC, following this pattern:

First hop: arrayA to arrayB
Second hop: arrayB to arrayA
Third hop: arrayA to arrayC
Fourth hop: arrayC to arrayA
Then the pattern starts over, continuing until the journey ends.

The rule to decide Gloria's move is: She uses the current element's value in the array as an index for her next array. For example, if Gloria is at arrayA[1]=2, she would move to arrayB[2].

The pattern repeats itself until one of the following occurs:

Gloria's path repeats by visiting a position in arrayB or arrayC that was already visited, indicating that she is stuck in a loop and cannot progress further, OR
Gloria tries to access an index that exceeds the length of an array (for example, attempting to access arrayA[4] when arrayA only contains 4 items indexed from 0 to 3), in which case Gloria's journey should also stop.

***Your Task***

Your task is to calculate the sum of the maximum values that Gloria encounters in arrayB and arrayC during her journey.

Each input array consists of n items, where n ranges from 1 to 100, inclusive. Every item in the arrays is a non-negative integer and falls within the range of 0 to 99, inclusive.

***EXAMPLE***

Consider arrayA = [2, 1, 3, 0], arrayB = [1, 3, 2, 4], and arrayC = [4, 2, 5, 1]. Gloria's journey would look like:

She begins at arrayA[0] = 2 which leads her to arrayB[2] = 2.
She then goes back to arrayA[2] = 3, and then to arrayC[3] = 1.
She returns to arrayA[1] = 1, then makes a hop to arrayB[1] = 3.
She goes back to arrayA[3] = 0 and then proceeds to arrayC[0] = 4.
Now Gloria would go to arrayA[4], however, since arrayA[4] doesn't exist because arrayA only contains 4 elements indexed from 0 to 3, Gloria's journey stops here.

During her journey, Gloria encounters the maximum value 3 in arrayB and 4 in arrayC. The function should return 7, the sum of these two maximum values.

***HINT***

1. The loop should stop if Gloria visits the same position (index) in an array again, rather than checking the values she finds.
2. Be careful with max() on empty lists, as it might cause an error if Gloria doesn't reach an array before the journey ends.


## Solution

```python
def solution(arrayA, arrayB, arrayC):
    indexA = 0
    indexBorC = None
    
    inArrayA = True
    inArrayB = False
    inArrayC = False
    nextB = True
    
    arrayB_value_visited = []
    arrayC_value_visited = []
    arrayB_index_visited = []
    arrayC_index_visited = []
    
    while True:
        
        if inArrayA:
            inArrayA = False
            if indexA >= len(arrayA):
                break
            indexBorC = arrayA[indexA]
            if nextB:
                inArrayB = True
                inArrayC = False
                nextB = False
            else:
                inArrayC = True
                inArrayB = False
                nextB = True
        
        if inArrayB:
            if indexBorC >= len(arrayB):
                break
                
            indexA = arrayB[indexBorC]
            if indexBorC in arrayB_index_visited:
                break
                
            arrayB_value_visited.append(indexA)
            arrayB_index_visited.append(indexBorC)
            inArrayA = True
        
        if inArrayC:
            if indexBorC >= len(arrayC):
                break
                
            indexA = arrayC[indexBorC]
            if indexBorC in arrayC_index_visited:
                break
                
            arrayC_value_visited.append(indexA)
            arrayC_index_visited.append(indexBorC)
            inArrayA = True
        
    return max(arrayB_value_visited) + max(arrayC_value_visited)

arrayA = [2, 1, 3, 0]
arrayB = [1, 3, 2, 4]
arrayC = [4, 2, 5, 1]
assert 7 == solution(arrayA, arrayB, arrayC)

arrayA = [2, 0, 1]
arrayB = [1, 3, 2]
arrayC = [2, 0, 1]
assert 2 == solution(arrayA, arrayB, arrayC)

arrayA = [1, 1, 0]
arrayB = [2, 1, 3]
arrayC = [2, 0, 1]
assert 1 == solution(arrayA, arrayB, arrayC)

arrayA = [0, 2, 0]
arrayB = [1, 1, 2]
arrayC = [0, 1, 2]
assert 3 == solution(arrayA, arrayB, arrayC)

arrayA = [1, 1, 2, 0]
arrayB = [2, 2, 1, 3]
arrayC = [1, 2, 3, 4]
assert 5 == solution(arrayA, arrayB, arrayC)
```

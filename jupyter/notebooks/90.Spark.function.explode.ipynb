## pyspark.sql.functions.explode()

explode(): Flattens nested structures by expanding an array or map into separate rows.

References:
- https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.functions.explode.html

### Exploding Column Containing a List

Exploding Column Containing a List

In PySpark, the explode() function transforms an array or map column into multiple rows. For each element in the array or each key-value pair in the map, a new row is created, with all other column values in that row being duplicated.

Key Behaviors:
- Data Duplication: All non-exploded columns in the row are replicated for every element found in the exploded column.
- Performance: Exploding large arrays significantly increases the number of rows in your DataFrame, which can impact memory and processing time.

See also:
- pyspark.sql.functions.posexplode()
- pyspark.sql.functions.explode_outer()
- pyspark.sql.functions.posexplode_outer()
- pyspark.sql.functions.inline()
- pyspark.sql.functions.inline_outer()


```python
# Explode column containing a list

from pyspark.sql import SparkSession
from pyspark.sql.functions import explode

spark = SparkSession.builder.appName("ExplodeExample").getOrCreate()

data = [(1, ["apple", "banana"]), (2, ["cherry"])]
df = spark.createDataFrame(data, ["id", "fruits"])

# Explode the 'fruits' array into multiple rows
exploded_df = df.select("id", explode("fruits").alias("fruit"))
exploded_df.show()
```

### Exploding Column Containing a Map

Exploding Column Containing a Map

Key Considerations
- Map Explosion: When used on a MapType column, it generates two default columns named key and value.
- Overriding The Default Column Names: If you want custom names for the exploded columns, you can use .alias("my_key", "my_value") after the explode function.
- Handling Nulls: The standard explode() function will remove rows where the map column is NULL or empty (like user_id 3 in the example).
- Preserving Nulls: If you want to keep rows with NULL or empty maps, use explode_outer() instead.
- Positional Explode: Use posexplode() if you need the specific index or order of the elements (though maps in Spark are typically unordered)


```python
# Explode column containing a map

from pyspark.sql import SparkSession
from pyspark.sql.functions import explode

# Initialize Spark Session
spark = SparkSession.builder.appName("ExplodeExample").getOrCreate()

# Sample data: User ID and a Map of {Product: Price}
data = [
    (1, {"Laptop": 1200, "Mouse": 25}),
    (2, {"Monitor": 300}),
    (3, None) # Null map example
]

df = spark.createDataFrame(data, ["user_id", "purchases"])

# Explode the map column
# It creates two default columns: 'key' and 'value'
exploded_df = df.select("user_id", explode("purchases"))

exploded_df.show()
```

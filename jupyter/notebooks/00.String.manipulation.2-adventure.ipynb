## Introduction

***Introduction***

You are provided with a string of n lowercase English characters, where n ranges from 1 to 500 inclusive. Your task is to return a dictionary where each key-value pair represents a letter k and its corresponding numerical representation v.

The numerical representation v of each character k is computed as follows: replace k with the character that comes three characters before it in the alphabetical order (wrap around to z when this is less than a), then multiply the ASCII value of the new character by the frequency of k in the provided string.

Your function should return a dictionary of the letters in the string and their corresponding numerical representations, sorted in ascending order by the characters.

The returned dictionary should be in the format: ***{'character': numerical_representation}***

***Hints***

- Each character's ASCII value can be obtained using Python's built-in ***ord function***
- The character corresponding to an ASCII value can be obtained using the ***chr function***.

***Example***

Given the string 'abc', your function should return: {'a': 120, 'b': 121, 'c': 122}
- replace 'a' with 'x' and multiply its ASCII value (120) by its frequency (1) to get 120.
- replace 'b' with 'y' and multiply its ASCII value (121) by its frequency (1) to get 121.
- replace 'c' with 'z' and multiply its ASCII value (122) by its frequency (1) to get 122.
- Then, we sort them based on the characters.

## Solution

```python
def solution(letters):
    # Form letter -> count dictionary
    letter_count_dict = {}
    for letter in letters:
        if letter in letter_count_dict.keys():
            letter_count_dict[letter] = letter_count_dict[letter] + 1
        else:
            letter_count_dict[letter] = 1
    print(f"letter_count_dict: {letter_count_dict}")

    # Form letter -> product dictionary
    letter_product_dict = {}
    for letter, count in letter_count_dict.items():
        letter_product_dict[letter] = count * calculate_value(letter)
    print(f"letter_product_dict: {letter_product_dict}")

    # Sort dictionary and return it
    sorted_dict = dict(sorted(letter_product_dict.items()))
    print(f"returning sorted_dict: {sorted_dict}")
    print()
    return sorted_dict
            
# Given a letter, calculate the value
def calculate_value(letter):
    unicode_d = ord("d")
    unicode = ord(letter)
    if (unicode < unicode_d):
        return wrap_around(letter)
    else:
        return unicode - 3

def wrap_around(letter):
    if letter == "a":
        return ord("x")
    if letter == "b":
        return ord("y")
    if letter == "c":
        return ord("z")

# Tests
assert solution("abc") == {'a': 120, 'b': 121, 'c': 122}
assert solution("def") == {'d': 97, 'e': 98, 'f': 99}
assert solution("xyz") == {'x': 117, 'y': 118, 'z': 119}
assert solution("aabccc") == {'a': 240, 'b': 121, 'c': 366}
assert solution("abcdefghij") == {'a': 120, 'b': 121, 'c': 122, 'd': 97, 'e': 98, 'f': 99, 'g': 100, 'h': 101, 'i': 102, 'j': 103}
assert solution("zzzzzzzzzz") == {'z': 1190}
assert solution("abcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabcabc") == {'a': 3600, 'b': 3630, 'c': 3660}
assert solution("xyzaaxz") == {'a': 240, 'x': 234, 'y': 118, 'z': 238}
```

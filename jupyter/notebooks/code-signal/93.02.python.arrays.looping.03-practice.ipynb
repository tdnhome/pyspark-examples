## Introduction

***Introduction***

- You are given a string that represents a sentence in which words are separated by spaces. 
- The input sentence consists of ASCII characters from the space character (' ') up to the tilde character ('~'), with its length ranging between 1 and 500, inclusive.
- These characters form words separated by spaces, without any consecutive space characters.

***Task***

Your task is to create a Python **function(sentence, c)** that
- **identifies and concatenates the second half of each word with an even number of characters**,
- ensuring the characters of this second half go before c, second input of the function, in the ASCII table.
- Then, combine these characters into a single string, maintaining the order in which they appear in the sentence.

***Example***

- Sentence: "Python is a high-level programming language." and the character "n".
- The word 'Python' consists of 6 characters (an even number), and the second half of this word is 'hon'.
- In this second half, only 'h' is less than 'n'.

The output of your function, in this case, should be: "h", as it's the only character that meets the conditions.

***Hints***

- For the character comparison ('<' character), use the ASCII values since all characters in the sentence are ASCII.
- ASCII codes for characters can be found by using Python's built-in function ord().


## Hint: Find ASCII value of a character

In Python, ord() and chr() are built-in functions that act as exact opposites to convert characters to their numeric Unicode code points and vice-versa. They provide the foundation for character arithmetic, text ciphering, and custom string validation.

```python
def ord_and_chr(c):
    order = ord(c)
    print(f"given:{c} -> ord:{order} ->chr:{chr(order)}")

ord_and_chr("a")
ord_and_chr("A")
```

## Solution

```python
# You are given a string that represents a sentence in which words are separated by spaces. 
# The input sentence consists of ASCII characters 
# These characters form words separated by spaces, without any consecutive space characters.
def solution(sentence, c):
    result = ""
    words = sentence.split(' ')
    for word in words:
        # confirms whether the length of the word is even
        if len(word) % 2 == 0:
            chars = filter_chars(word, c)
            result += chars
    
    print(f"result: {result}")
    return result

# Identifies and concatenates the second half of each word with an even number of characters,
# ensuring the characters of this second half go before upper_bound, the second input of the function, in the ASCII table.
# Then, combine these characters into a single string, maintaining the order in which they appear in the sentence.
def filter_chars(word, upper_bound):
    start = int(len(word)/2)
    end = len(word)
    upper_bound_ord = ord(upper_bound)

    result = ""
    for i in range(start, end):
        c = word[i]
        if ord(c) < upper_bound_ord:
            result += c
    return result


# Test from the example section
assert solution("Python is a high-level programming language.", 'n') == "hleel"

# Test from https://codesignal.com/learn/course/93/unit/2/practice/3
assert solution("Practice makes perfect.", 'f') == "ceec."
assert solution("Mastering Advanced Looping and Implementation.", 'l') == "ced"
assert solution("I will pass this test!", 'w') == "llssis"
assert solution("Participate in exciting challenges.", 'a') ==  ""
assert solution("The quick brown fox jumps over the lazy dog.", 'f') == "e."
assert solution("This sentence is a test sentence.", 't') == "isencess"
assert solution("This is a long string with lots of characters and words.", 'v') == "issngingthtsfctersds."
assert solution("Another sentence to test the function.", 'o') == "ence"
assert solution("Some additional variety in test cases.", 'k') == "eiae."
```

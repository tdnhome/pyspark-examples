## Problem Statement

***Introduction***

We're going to explore a fascinating scenario involving a wonderful small town, its houses, and a fun balloon game. 

***Task Statement***

Picture a quaint, small town where every house is numbered sequentially from 1 to n. One day, a festive town event is held, and a random number of balloons are tied to each house. 

A fun game is played: at each step of the game, each house sends half of its balloons to the neighboring house simultaneously (the neighbor on the right side, and for the last house, the neighbor is the first house). The game goes on until at some step there are no changes in the amount of balloons compared to the previous step.

Create a Python function, solution(balloons), where balloons is a list representing the number of balloons at each house. The function should simulate this game and return the number of steps in the game.

***Hints***

1. When the number of balloons x is odd, than the house sends (x - 1) / 2 balloons.
2. It's essential to note that we're dealing with a cyclical event. In other words, when iterating over our balloons array, we need to perceive the array as circular, meaning balloons[n - 1] should refer back to balloons[0]. This concept of cyclicity becomes crucial when we consider the last house passing balloons to the first.


***Example***

If balloons = [4, 1, 2], the output should be solution(balloons) = 3. 
- After the first step, the list becomes [3, 3, 1]. This is because
  - the first house sends 2 balloons and gets 1,
  - the second house sends nothing but gets 2, and
  - the third house sends 1 but receives nothing.
- After the second step, the list becomes [2, 3, 2] and never changes after that. So after the third step, the process finishes.

## Solution

```python
# Step 1 - Setting Up The Loop
# First, we need to set up a loop to iterate the rounds of the balloon sharing. 
# This loop should continue as long as the list changes.

def solution(balloons):
    steps = 0   # initialize the result
    
    while True: # loop until no change on the number of balloons on all houses from previous loop 
        steps += 1
        new_balloons = balloons.copy()  # Store updated balloon counts

# Step 2 - Sharing Balloons
# Throughout each cycle, each house must share half of its balloons with the next house.
# We must also ensure that the last house shares balloons with the first house at the end 
# of each cycle — for this, we'll use the handy modulo % operator: find the remainder of the division
        
        for i in range(len(balloons)):
            share = balloons[i] // 2  # floor division operator: divide two numbers and round the result down
            new_balloons[i] -= share  # Decrease balloons of current house.
            new_balloons[(i + 1) % len(balloons)] += share  # Increase balloons of next house.

# Step 3 - Check if the solution is found: if o change on the number of balloons on all houses from previous loop
        if new_balloons == balloons:
            break

# Step 4 - Prepare for next loop: Update balloons with new counts
        balloons = new_balloons 
        
    return steps

# Test case from the example section
# If balloons = [4, 1, 2], the output should be solution(balloons) = 3. 
balloons = [4, 1, 2]
expected = 3
result = solution(balloons)
print(f"Result: {result}")
assert result == expected
```

## Exercise: Checking the list

REF: https://codesignal.com/learn/course/90/unit/2/practice/4

Checking your packing list is essential. 
- **'in' operation**: Check whether an item is on your packing list
- **list.'index'**: determine its position.

```python
# Travel Packing List and Selection
packing_list = ['clothes', 'toothbrush', 'passport', 'camera']
item_to_check = 'passport'

# TODO: Write a line of code to check if the item_to_check is in the packing_list
# TODO: Find the index of item_to_check in the list if it is packed, otherwise set it to -1
is_item_packed = item_to_check in packing_list
item_index = -1
if is_item_packed:
    item_index = packing_list.index(item_to_check)

# Print out the results
print("Is the item packed?", is_item_packed)
print("Item index:", item_index)
```

## Exercise: String manipulation

REF: https://codesignal.com/learn/course/90/unit/2/practice/6

You're given a mission name (a string), and your task is to change its first and last characters. Show the original first and last characters, make the changes, and then display the updated string. Remember, strings are immutable, so you'll need to find a creative way to achieve your goal!

```python
# Given mission name
mission_name = "Sort"

# TODO: Print the first and the last character of the mission name
print(mission_name[0])
print(mission_name[-1])

# TODO: The mission needs a minor update. We aim to change its first letter to 'P' and the last letter to `k`, obtaining the word "Pork".
# Remember, strings in Python are immutable, so you cannot alter them directly.

mission_name = "P" + mission_name[1:]
mission_name = mission_name[:-1] + "k"

# TODO: Print the updated mission name
print(mission_name)
```

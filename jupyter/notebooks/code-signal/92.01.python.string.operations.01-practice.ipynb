***Task***
- Given a string of n characters, with n varying from 1 to 1000, inclusive. 
- Your task is to write a Python function that takes this string as input, applies the following operations, and finally returns the resulting string.
  - Split the given string into individual words, using a space as the separator.
  - Convert each word into a list of its constituent characters, and shift each list once to the right (with the last element moving to the first position).
  - After the rotations, reassemble each word from its list of characters.
  - Join all the words into a single string, separating adjacent words with a single space.
  - Return this final string as the function's output.

***Constraints***
- The input string will neither start nor end with a space, nor will it have multiple consecutive spaces.
- Each word will contain only alphabets and digits, and its length will range from 1 to 10.
- The words are case-sensitive; for example, 'word' and 'Word' are considered distinct.
- Your program should output a single string with the words rotated by their lengths while preserving their original order.

***Example***
- Input string: "abc 123 def"
- Expected result: "cab 312 fde"


```python
def solution(s):
    words = s.split()
    new_words = [''.join(shift_chars_right(word)) for word in words]
    result = " ".join(new_words)
    print(f"Result: {result}")
    return result


def shift_chars_right(s):
    last_item_index = len(s) - 1
    return s[last_item_index] + s[0:last_item_index]


# Test cases from https://codesignal.com/learn/course/92/unit/1/practice/1
assert solution("abc 123 def") == "cab 312 fde"
assert solution("abc 123 def ghi") == "cab 312 fde igh"
assert solution("bat") == "tba"
assert solution("raceCar") == "rraceCa"
assert solution("mAnGo666 TaCo123i") == "6mAnGo66 iTaCo123"
assert solution("_ab 77Y UwF88") == "b_a Y77 8UwF8"
assert solution("SingleWord") == "dSingleWor"
assert solution("abcdefghij") == "jabcdefghi"
assert solution("ZzZzZzZ 1234567890 zYxWvUtS") == "ZZzZzZz 0123456789 SzYxWvUt"
```

### For Loop: Looping Over Collections


Python loops can directly work with collections, making it easy to handle elements of a list or string.

#### Looping Over List

````python
friends = ['Alice', 'Bob', 'Charlie', 'Daniel']
# `friend` is the loop variable, taking each name in the `friends` list
for friend in friends:
    # for each `friend`, this line is executed
    print(f"Hello, {friend}! Nice to meet you.")
````

#### Looping Over Strings

````python
word = 'hello'
# `letter` is each individual character in the `word`
for letter in word:
    print(letter) # prints each letter in 'hello'
````

#### Looping over a range(start, end, step)

In Python, a for loop works with any sequence, like lists or strings. Let's print a range of numbers using a for loop:

In the following example, **num** runs through each number in the range of 5 (from 0 to 4)

```
for num in range(5):
    # This line will print numbers from 0 to 4
    print(num)
```
Each loop iteration updates the variable (num) to the next sequence value before executing the code block.

````python
for num in range(5):
    print(num)
````

### While Loop: Loop while something is true (loop until something is false)

While loops execute code continuously until a condition becomes false. Here's a simple example:

The loop keeps running ***until num is*** greater than or equal to 5

```
num = 0
while num < 5:
    print(num)
    # increase num by 1 each iteration
    num += 1
```
As you can see, before each iteration, Python checks the condition (num < 5). If it's true, the code block runs; otherwise, Python exits the loop.



````python
num = 0
while num < 5:
    print(num)
    num += 1
````

### Applications of Looping

Loops are integral to programming. They are used in various programming applications, such as summing numbers in a list or parsing text.

#### Applications with lists

````python
# List of numbers
numbers = [1, 2, 3, 4, 5]
total = 0
# `num` is each number in `numbers`
for num in numbers:
    total += num # add each number in the list
print(total) # prints the total sum
````

#### Applications with strings

````python
# string
text = 'hello'
vowel_count = 0
# `letter` is each character in `text`
for letter in text:
    # If a vowel letter is found, increment the count
    if letter in 'aeiou':
        vowel_count += 1
print(vowel_count) # prints the count of vowels
````

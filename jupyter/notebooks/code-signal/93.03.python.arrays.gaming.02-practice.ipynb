## Introduction

***Introduction***

Your task is to design a 1-dimensional game where a player moves along a path determined by an array of integers.

- The path is an array of integers, each ranging from -100 to 100, inclusive.
- The size of the array n, i.e., the total number of steps on the path, can range from 1 to 500, inclusive.
- Each integer a_i in the array signifies how many steps the player can move and in which initial direction:
  - A positive integer allows the player to move that many steps to the right.
  - A negative integer directs the player to move that many steps to the left.
  - Zero signifies a blockade that prevents further movement.
- The game proceeds along the following rules:
  - The player starts at the first position of the array (0-indexed) and moves according to the value at the player's current position in the array.
  - If the value in the current position is zero, then the game ends.
  - If the player's current position leads them outside of the array's boundaries, then the player reverses their direction and continues to move according to similar rules, but now the directions are inverted: positive integers lead the player to the left, and negative integers point to the right.
  - The game ends when the player encounters a blockade or the array boundaries for the second time and so can no longer move.

***Task***

You are to implement a function titled evaluatePath(numbers). This function should take an array of integers as input, representing the path and its rules, and return a tuple (position, moves), where:
- position: This is the player's final position (0-indexed) when the game ends.
- moves: This is the total number of moves made by the player until the game ends.

It's guaranteed that the game will not lead to an infinite loop, i.e. the path to the next blockade or the array boundaries would not require the player to visit the same position more than once.

***Example***

- array = [3, 4, 1, 1, -3, 1]
- output (4, 5)
- Reasoning:
  - At position 0, the value is 3. They move 3 steps to the right, landing on the 3rd position. Total moves till now: 1.
  - At position 3, the value is 1. They move 1 step to the right, landing on the 4th position. Total moves till now: 2.
  - At position 4, the value is -3. They move 3 steps to the left, landing back on position 1. Total moves meanwhile: 3.
  - At position 1, the value is 4. They move 4 steps to the right, landing on position 5. Total moves thus far: 4.
  - At position 5, the value is 1, which would lead them out of the array's right boundary. ***So, they reverse their direction***.
  - With the reversed direction at position 5, they move 1 step to the left and land on position 4. Total moves till now: 5.
  - Now, with the reversed direction, the player is at position 4 where the value is -3. In the reversed direction, -3 indicates 3 steps to the right. But this would again lead to the right boundary of the array. ***Since they have already reversed direction once***, they cannot move further in any direction and the game ends.
  - At the end of the game, the player is at position 4 having made a total of 5 moves, thereby, the function returns (4, 5).

```python
def evaluatePath(numbers):

    # Keep track of moving direction: 1=right or forward, -1=left or backward
    moving_direction = 1 
    
    # Keep track of number of time the reversed_direction happened
    reversed_direction = 0 

    # Keep track of the results
    total_moves = 0
    current_position = 0

    # The max # of moved can be from 0 to the end of the range and reversed direction back to 0
    for i in range(len(numbers) * 2):
        
        value = numbers[current_position]
        # print(f"At position {current_position}, the value is {value}, total_moves = {total_moves}, moving_direction: {moving_direction}, reversed_direction: {reversed_direction}")

        # If the value in the current position is zero, then the game ends.
        if value == 0:
            break

        # Calculate temporary position
        # The steps (either backward or forward) is calculated as (value_at_position_i) times (moving_direction)
        steps = numbers[current_position] * moving_direction
        new_position = current_position + steps

        # Check if the temporary position would go out of bound
        if new_position >= len(numbers) or new_position < 0:

            # Only one time of reversed direction is allow
            if reversed_direction > 0:
                print("Change direction is no longer allowed")
                break

            # Recalculate the position with the new moving direction
            print(f"Change direction from {moving_direction}")
            reversed_direction += 1
            moving_direction = moving_direction * -1

            # Calculate temporary position for the new direction and check again
            steps = numbers[current_position] * moving_direction
            new_position = current_position + steps
            if new_position >= len(numbers) or new_position < 0:
                print("Change direction is no longer allowed")
                break
            else:
               current_position = new_position
        else:
            current_position = new_position
            
        total_moves += 1

    result = (current_position, total_moves)    
    print(f"Result: {result}")
    return result


# Test case from example on the introdution section
assert evaluatePath([3, 4, 1, 1, -3, 1]) == (4, 5)

# Test cases from https://codesignal.com/learn/course/93/unit/3/practice/2
assert evaluatePath([2, 1, -3, 4]) == (2, 1)
assert evaluatePath([0]) == (0, 0)
assert evaluatePath([3, 4, 1, 1, -3, 1]) == (4, 5)
assert evaluatePath([3, -3]) == (0, 0)
assert evaluatePath([3, 2, -1, 2, 2, -1, 4]) == (1, 7)
assert evaluatePath([1]*500) == (0, 998)
```

***Task***
- Your task is to implement a function that duplicates every digit in a given non-negative integer number, n.
- For example, if n equals 1234, the function should return 11223344.
- To prevent possible integer overflow, it is guaranteed that n will be a non-negative integer that does not exceed 10**4
- Solve this task without converting n into a string or performing any other type of casting. Your job is to work strictly with integer operations.

***Hints***
- Focus on the essence of the problem, which is processing each digit of the number independently while maintaining the digit order.
- There is no need to look for mathematical patterns or clever simplifications; plain and straightforward processing will suffice.
- Utilize the toolbox of basic programming skills: loops, conditions, and mathematical operations.


```python
def solution(n):
    result = 0     # Step 1
    s = 1
    while n > 0:   # Step 2
        # Step 3
        digit = n % 10                # Extract the last digit of n
        result = (digit * s) + (digit * s * 10) + result # Calculate the new result
        n = n // 10  # We are done with the last digit, remove it and continue next loop
        s = s * 100  # Recalculate significant value. Since each digit is double, the digit's signficant is also double
    print (f"Result: {result}")
    return result


# Test cases from https://codesignal.com/learn/course/91/unit/2/practice/2
assert solution(1234) == 11223344 # From example in the introduction section
assert solution(1) == 11
assert solution(22) == 2222
assert solution(9876) == 99887766
assert solution(10000) == 1100000000
assert solution(0) == 0
assert solution(3333) == 33333333
assert solution(4444) == 44444444
assert solution(5555) == 55555555
assert solution(6666) == 66666666
```

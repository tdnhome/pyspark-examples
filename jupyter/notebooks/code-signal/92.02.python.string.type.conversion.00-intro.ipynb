## Task

***Task***
- The task du jour involves creating a Python function named parse_and_multiply_numbers().
- The function will take a string as input. This input string is quite special — it will have numbers and words jumbled together in a free-spirited manner.
- The function's job is to parse this input string, find all the numbers, convert these numbers (which are currently strings) into integer data types, subsequently multiply all these numbers together. The output? It’s the product of all those numbers!

***Example***
Input: "I have 2 apples and 5 oranges,"
Output: The product of 2 and 5, which is 10.

## Solution Building Steps

***Step 1 - Initialization***
- The first course of action is to parse the string and capture the numbers. But how do we do this? We can sequentialize our search.
- Let's create an empty string num where we will gather digits and an empty list numbers to collect all the numbers found:
```
input_string = "I have2apples and5oranges"
num = ""
numbers = []
```

***Step 2 - Checking if a string is a number using str.isdigit() method***
- The next move is to iterate through the input string character by character.
- If we come across a digit, we add it to our num string.
- If the character is not a digit, and num is not empty, it means we've reached the end of a number.
```
for char in input_string:
    if char.isdigit():
        num += char
    elif num:
        numbers.append(int(num))
        num = ""
print(numbers)
```
After running this code, the output on the console will be [2, 5].

***Step 3 - Multiply all the numbers in the list numbers together***
- Lastly, we multiply all the numbers in the list numbers together.
- This multiplication is saved in the variable result.
```
result = 1
for number in numbers:
    result *= number
print(result)
```

## Code

````python
def parse_and_multiply_numbers(input_string):
    num = ""
    numbers = []
    for char in input_string:
        if char.isdigit():
            num += char              # keep adding digit to num for the current sequence
        elif num:
            numbers.append(int(num)) # save the number to the list
            num = ""                 # reset for the next sequence
    if num:
        numbers.append(int(num))     # last check

    # multiply all the numbers in the list numbers together.
    result = 1
    for number in numbers:
        result *= number
    print(f"Result: {result}")
    return result

assert parse_and_multiply_numbers("I have2apples and5oranges") == 10
````

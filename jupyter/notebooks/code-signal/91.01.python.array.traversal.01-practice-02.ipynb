***Task***

- You are provided with a list of n integers, where n ranges from 2 to 200, inclusive. 
- The task is to return a list of tuples, each containing a pair of an integer and its reverse counterpart.
- In this context, the reverse counterpart of a number is the number with its digits reversed. For example, the reverse counterpart of 123 is 321.
- Your output should be a list of tuples with the original numbers and their reverse counterparts. The integers in the given list will range from 1 to 99999, inclusive.

***Hints***

- You must iterate through the list to find the reverse counterpart for each integer. 
- If this reversed number exists in the list, create a tuple with the original number and its reverse counterpart. If not, skip it.
- The reverse counterpart of a single digit number is the same number.
- For numbers that result in leading zeros when reversed (e.g., 100 becomes 001, which is 1), consider only the integer value (i.e., 1).
- It's guaranteed that the original list will not contain integers with leading zeros.

***Example***
- numbers = [12, 21, 34, 43, 56, 65], the output should be [(12, 21), (21, 12), (34, 43), (43, 34), (56, 65), (65, 56)].

***Example of special case***
- numbers = [12, 55], the output should be [(55, 55)].
  - For 12, the reverse is 21. Since 21 is not in the original list, it is not included.
  - For 55, the reverse is 55. Since 55 is in the original list, it is included.


```python
def solution(numbers):
    result = []
    
    # Iterate through the list to find the reverse counterpart for each integer
    for i in range(len(numbers)):
        next_pair = reverse_number(numbers[i])
        # If this reversed number exists in the orgininal list, create a tuple with the original number and its reverse counterpart. 
        n,r = next_pair
        if r in numbers:
            result.append(next_pair)
            
    print(f"Result: {result}")
    return result


def reverse_number(number):
    s = str(number)
    result = ""
    for i in range(len(s)):
        result = s[i] + result
        
    return (number, int(result))


assert solution([12, 21, 34, 43, 56, 65]) == [(12, 21), (21, 12), (34, 43), (43, 34), (56, 65), (65, 56)]
assert solution([12, 21, 34, 43, 56, 65]) == [(12, 21), (21, 12), (34, 43), (43, 34), (56, 65), (65, 56)]
assert solution([10, 1, 20, 2]) == [(10, 1), (1, 1), (20, 2), (2, 2)]
assert solution([1234, 4321]) == [(1234, 4321), (4321, 1234)] 
assert solution([1111, 1111]) == [(1111, 1111), (1111, 1111)]
assert solution([12345, 54321, 11111, 44444, 22222, 88888]) == [(12345, 54321), (54321, 12345), (11111, 11111), (44444, 44444), (22222, 22222), (88888, 88888)]

# Special case: the reverse of 12 is 21 which is not in the original list, so the tuple (12, 21) is not included
assert solution([12, 55]) == [(55, 55)]
```

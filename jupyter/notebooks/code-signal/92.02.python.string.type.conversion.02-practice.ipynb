***Task***
- You are given a string s of length n, with n ranging from 1 to 500 inclusive.
- This string represents the complex and jumbled record of a sports game.
- It combines player names and scores but lacks a uniform structure.
- The player names consist of words made up of lowercase English alphabets (a-z), while the scores are integers ranging from 1 to 100 inclusive.
- Your mission involves writing a Python function solution(). This function should parse the given string, isolate the integers representing player scores, and return the sum of these scores.

***Example***
- Input: "joe scored 5 points, while adam scored 10 points and bob scored 2, with an extra 1 point scored by joe"
- Result: the sum 5 + 10 + 2 + 1, which totals 18

```python
def solution(s):
    words = s.split()
    result = 0    
    for word in words:
        result += find_score(word)
    print(f"Result: {result}")
    return result


def find_score(word):
    s = extract_number(word)
    if s.isdigit():
        return int(s)
    else:
        return 0

def extract_number(word):
    length = len(word)
    result = ""
    for i in range(length):
        if word[i].isdigit():
            result += word[i]
    return result
    
# Test cases from https://codesignal.com/learn/course/92/unit/2/practice/2
assert solution("joe scored 5 points, while adam scored 10 points and bob scored 2, with an extra 1 point scored by joe") == 18
assert solution("michael scored 100 points") == 100
assert solution("lena scored 50 points and lee scored 50 points") == 100
assert solution("sam scored 25 points, john scored 25 points, jim scored 25 points, and sue scored 25 points") == 100
assert solution("1 point scored by max") == 1
assert solution("no points scored in this game") == 0
assert solution("abc scored 3 points and def scored 9 points then ghi scored 27 points") == 39
assert solution("game score: pete 2 points, eve 4 points, zane 8 points") == 14
assert solution("jake scored1point, john scored2points") == 3
assert solution("this game ended with no score") == 0
```

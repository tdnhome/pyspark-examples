***Task***
- Given a string consisting of words separated by whitespace
- Your task is to write a Python function that accepts this string.
- It then replaces each character in the words with the corresponding character opposite in the English alphabet and stitches them all together to form a new string.
- The opposite letter mappings are as follows: a ↔ z, b ↔ y, c ↔ x, ..., m ↔ n, n ↔ m, ..., x ↔ c, y ↔ b, z ↔ a.
- The mapping is case-sensitive.

***Constraints***
- The input string will include between 1 and 100 words.
- Each word consists of characters separated by white space.
- A word is composed of characters ranging from a to z or A to Z.
- If a word contains a lowercase 'a', for instance, it should be replaced with 'z', 'b' with 'y', 'c' with 'x', and so on, maintaining the same case.
- For words with an uppercase 'A', it should be replaced with 'Z', 'B' with 'Y', 'C' with 'X', and so forth, while preserving the uppercase.
- The given string will not start or end with a space, and there will be no occurrence of double spaces.
- After transforming the characters of the words, form a new string by taking the last word first and appending the remaining words in their original order, each separated by spaces.

***Example***
- Input: "CapitaL letters"
- Expected result: "ovggvih XzkrgzO"

***Hints***
- ord(): a built-in method that converts a single Unicode character into its corresponding integer Unicode code point
- chr(): a built-in method that takes a Unicode code point as an integer and returns the corresponding character string
- To find the "opposite" character in the English alphabet (e.g., a↔z, b↔y), you can use Python's ord() and chr() functions
- Observations: the sum of the ASCII values of opposite pairs (like 'A' and 'Z' or 'a' and 'z') is constant
  - 65 + 90 = 155 for uppercase
  - 97 + 122 = 219 for lowercase
- Formulas:
  - chr(219 - ord(c)): for lowercase range a-z
  - chr(155 - ord(c)): for uppercase range A-Z

```python
def solution(s):
    words = s.split()

    # create a list of words, each having opposite letters, from words in the input sentence
    new_words = [''.join(get_opposite_word(word)) for word in words]
    
    # pop the last word out of the list and insert it at the beginning of the list
    last_word = new_words.pop()
    new_words.insert(0, last_word)

    # create a sentence from the word list 
    result = " ".join(new_words)
    
    print(f"Result: {result}")
    return result


def get_opposite_word(word):
    opposite_word = ""
    for i in range(len(word)):
        opposite_word += get_opposite_char(word[i])
    return opposite_word


def get_opposite_char(char):
    if 'a' <= char <= 'z':
        return chr(219 - ord(char))
    elif 'A' <= char <= 'Z':
        return chr(155 - ord(char))
    return char


# Test cases from https://codesignal.com/learn/course/92/unit/1/practice/2
assert solution("CapitaL letters") == "ovggvih XzkrgzO"
assert solution("Hello") == "Svool"
assert solution("ABC") == "ZYX"
assert solution("abc") == "zyx"
assert solution("A quick brown FOX jumps over the lazy DOG") == "WLT Z jfrxp yildm ULC qfnkh levi gsv ozab"
assert solution("Zebra") == "Avyiz"
assert solution("CapitaL letters") == "ovggvih XzkrgzO"
assert solution("loWer letters") == "ovggvih olDvi"
assert solution("OPPOSITE letters") == "ovggvih LKKLHRGV"
assert solution("An apple a day keeps the doctor away") == "zdzb Zm zkkov z wzb pvvkh gsv wlxgli"
assert solution("m n") == "m n"
```

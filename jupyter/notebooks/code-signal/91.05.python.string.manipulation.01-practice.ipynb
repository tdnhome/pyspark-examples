***Task***
- Run-length encoding is a simple form of data compression where sequences of data entities that are the same are stored as a single data entity along with its count. Each count must immediately follow the character it is associated with.
- Function:
  - name: encode_rle.
  - input: a string as an input argument
  - return: a new string that represents the input's run-length encoding.
  - The function should operate only on alphanumeric characters (numbers 0-9 and uppercase and lowercase letters A-Z, a-z).
  - For any other types of characters in the string, simply ignore them and do not include them in the final encoded output.

***Example***
- Input: "aaabbcccdde"
- Expected Output: "a3b2c3d2e1"

***Example***
- If the input string includes non-alphanumeric characters, such as "aaa@@bb!!c#d**e"
- The output should be "a3b2c1d1e1".


```python
def encode_rle(s):
    groups = ""
    current_group_char = None
    current_group_length = 0

    for char in s:
        if char.isdigit() or char.isalpha():
            if char == current_group_char:
                current_group_length += 1
            else:
                if current_group_char is not None:
                    groups += current_group_char
                    groups += str(current_group_length)
                current_group_char = char
                current_group_length = 1
    if current_group_char is not None:
        groups += current_group_char
        groups += str(current_group_length)

    print(f"Result: {groups}")
    return groups
    
assert encode_rle("aabbbcceddf") == "a2b3c2e1d2f1"
assert encode_rle("aaa@@bb!!c#d**e") == "a3b2c1d1e1"
assert encode_rle("AA111bbbc") == "A213b3c1"
assert encode_rle("a") == "a1"
assert encode_rle("AAABCC@@@D123df#$@# adedfeee333!!!!!FFFFFFF") == "A3B1C2D1112131d1f1a1d1e1d1f1e333F7"
assert encode_rle("abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789") == \
    "a1b1c1d1e1f1g1h1i1j1k1l1m1n1o1p1q1r1s1t1u1v1w1x1y1z1A1B1C1D1E1F1G1H1I1J1K1L1M1N1O1P1Q1R1S1T1U1V1W1X1Y1Z101112131415161718191"
assert encode_rle("") == ""
assert encode_rle("1") == "11"
assert encode_rle("11111111112222222222aaaaaaaaaaa") == "110210a11"
assert encode_rle("AaBbCcDdEeFfGgHhIiJjKkLlMmNnOoPpQqRrSsTtUuVvWwXxYyZz") == \
    "A1a1B1b1C1c1D1d1E1e1F1f1G1g1H1h1I1i1J1j1K1k1L1l1M1m1N1n1O1o1P1p1Q1q1R1r1S1s1T1t1U1u1V1v1W1w1X1x1Y1y1Z1z1"
        
```

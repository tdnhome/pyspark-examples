***Task***
- You are given a string filled with words.
- Your task is to write a Python function that takes this string as input.
- Your function should transform each word so that it follows this rule:
  - If the word starts with a letter, make the first character uppercase and make the rest of the characters lowercase.
  - If the word starts with a non-letter character (such as a number or an underscore), leave that first character unchanged and make the rest of the characters lowercase.
  - Finally, recombine the transformed words into a new string in the same order.

***Constraints***
- The input string will contain between 1 and 100 words.
- Each word is a sequence of characters separated by white space.
- Words consist of characters ranging from a to z, A to Z, 0 to 9, or even an underscore _.
- The provided string will not start or end with a space, and it will not contain double spaces.
- After transforming each word, the program should return a single string in which the words maintain their original order.
- If punctuation marks are attached to words (such as "Hello," or "world!"), keep the punctuation attached and in the same position.
- You are not required to separate punctuation marks from the words; just apply the same transformation to the whole word.

***Example***
- Input: "SoME rAndoM _TeXT"
- Output: "Some Random _text"t

***Example***
- Input: "hELLo, wORLD!"
- Output: "Hello, World!"

***Hints***
- isalpha(): is a string's method to check if the string is alphabet (not including numbers and special characters)

```python
def solution(s):
    words = s.split()

    # create a list of words, after transform each of them, from words in the input sentence
    new_words = [''.join(transform_word(word)) for word in words]
    
    # create a sentence from the word list 
    result = " ".join(new_words)
    
    print(f"Result: {result}")
    return result


def transform_word(word):
    '''
    If the word starts with a letter, make the first character uppercase and make the rest of the characters lowercase.
    If the word starts with a non-letter character (such as a number or an underscore), leave that first character unchanged and make the rest of the characters lowercase.
    '''
    result = ""
    first_char = word[0]
    if first_char.isalpha():
        first_char = first_char.upper()
    remaining_char = word[1:]
    return first_char + remaining_char.lower()


# Test cases from https://codesignal.com/learn/course/92/unit/1/practice/3
assert solution("hello world") == 'Hello World'
assert solution("HELLO WORLD") == 'Hello World'
assert solution("123 hello") == '123 Hello'
assert solution("_underscore") == '_underscore'
assert solution("single") == 'Single'
assert solution("Hello neat pythonistas_123") == 'Hello Neat Pythonistas_123'
assert solution("SoME rAndoM _TeXT") == 'Some Random _text'
assert solution("CAPS lock IS on") == 'Caps Lock Is On'
assert solution("mIxEd CaSe sample") == 'Mixed Case Sample'
assert solution("first second third fourth fifth sixth seventh eights ninth tenth") == 'First Second Third Fourth Fifth Sixth Seventh Eights Ninth Tenth'

```

***Task***
- Write a Python function that takes in a string and identifies all the consecutive groups of identical characters within it, with the analysis starting from the end of the string rather than from its beginning.
- A group is defined as a segment of the text where the same character is repeated consecutively.
- Function should return a list of tuples. Each tuple will consist of the repeating character and the number of its repetitions. For instance, if the input string is "aaabbcccdde", the function should output: [('e', 1), ('d', 2), ('c', 3), ('b', 2), ('a', 3)].

```python
def solution(s):
    groups = []
    current_group_char = None
    current_group_length = 0
    start = len(s) - 1 # inclusive
    stop = -1          # exclusive
    step = -1

    for i in range(start, stop, step):
        char = s[i]
        if char == current_group_char:
            current_group_length += 1
        else:
            if current_group_char is not None:
                groups.append((current_group_char, current_group_length))
            current_group_char = char
            current_group_length = 1
    if current_group_char is not None:
        groups.append((current_group_char, current_group_length))

    print(f"Result: {groups}")
    return groups

# Test cases from https://codesignal.com/learn/course/91/unit/5/practice/2
assert solution("aaabbcccdde") == [('e', 1), ('d', 2), ('c', 3), ('b', 2), ('a', 3)]
assert solution("aaaaaaaZ") == [('Z', 1), ('a', 7)]
assert solution("a") == [('a', 1)]
assert solution("abc12321cba") == [('a', 1), ('b', 1), ('c', 1), ('1', 1), ('2', 1), ('3', 1), ('2', 1), ('1', 1), ('c', 1), ('b', 1), ('a', 1)]
assert solution("123454321") == [('1', 1), ('2', 1), ('3', 1), ('4', 1), ('5', 1), ('4', 1), ('3', 1), ('2', 1), ('1', 1)]
assert solution("ABBA") == [('A', 1), ('B', 2), ('A', 1)]
assert solution("Radar") == [('r', 1), ('a', 1), ('d', 1), ('a', 1), ('R', 1)]
assert solution("$$$$$") == [('$', 5)]
assert solution("Rotor") == [('r', 1), ('o', 1), ('t', 1), ('o', 1), ('R', 1)]
assert solution("Red roses run no risk, sir, on Nurse's order") == [('r', 1), ('e', 1), ('d', 1), ('r', 1), ('o', 1), (' ', 1), ('s', 1), ('\'', 1), ('e', 1), ('s', 1), ('r', 1), ('u', 1), ('N', 1), (' ', 1), ('n', 1), ('o', 1), (' ', 1), (',', 1), ('r', 1), ('i', 1), ('s', 1), (' ', 1), (',', 1), ('k', 1), ('s', 1), ('i', 1), ('r', 1), (' ', 1), ('o', 1), ('n', 1), (' ', 1), ('n', 1), ('u', 1), ('r', 1), (' ', 1), ('s', 1), ('e', 1), ('s', 1), ('o', 1), ('r', 1), (' ', 1), ('d', 1), ('e', 1), ('R', 1)]
```

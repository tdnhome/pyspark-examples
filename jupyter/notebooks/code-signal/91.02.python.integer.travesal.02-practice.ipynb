***Task***
- Your task is to construct a function that accepts an integer n and returns the integer with the same digits as n, but in reverse order.
- You should implement your solution using a while loop.
- For instance, if the input is 12345, the output should be 54321.
- Keep in mind that n will always be a positive integer between 1 and 10**8

***Notes***
- Do not use built-in functions that convert the integer to another data type, such as a string, to reverse it.
- Solve the problem purely using mathematical operations and loop constructs.

```python
def solution(n):
    result = 0     # Step 1
    s = find_significant(n)
    while n > 0:   # Step 2
        # Step 3
        digit = n % 10                # Extract the last digit of n
        result = result + (digit * s) # Calculate the new result
        n = n // 10  # We are done with the last digit, remove it and continue next loop
        s = s // 10  # Recalculate the significant value
    print (f"Result: {result}")
    return result


def find_significant(value):
    s = 1
    while value > 10:
        s = s * 10           # calculate new significant value
        value = value // 10  # reduce the value
    return s


# Test cases from https://codesignal.com/learn/course/91/unit/2/practice/2
assert solution(12345) == 54321 # From example in the introduction section
assert solution(1) == 1
assert solution(54321) == 12345
assert solution(87654321) == 12345678
assert solution(56) == 65
assert solution(4567890) == 987654
assert solution(98765) == 56789
assert solution(876) == 678
assert solution(23456789) == 98765432
```

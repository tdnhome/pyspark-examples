***Task***
- You are tasked with writing a function that takes a positive integer, n, as input
- The function returns the number of times a digit appears immediately next to an identical digit in the number.
- In other words, your function should count how many times two equal digits appear consecutively when reading the number from left to right.

***Example***
- For n = 113224, there are two instances where a digit is immediately followed by the same digit: 11 and 22. Therefore, the output should be 2.
- For n = 444, the output should be 2, as there are two places where a digit is immediately followed by the same digit (44 and another 44).

***Hints***
- Keep in mind that n will be a positive integer ranging from 1 to 10**8, inclusive.
- You are not permitted to convert the number into a string or any other iterable structure for this task. You should work directly with the number.

```python
def solution(n):
    result = 0
    last_digit = None
    while n > 0:
        digit = n % 10 # Extract the last digit of n
        if digit == last_digit:
            result += 1
        else:
            last_digit = digit
        n = n // 10 # Done with the last digit, remove it and continue next loop
    print (f"Result: {result}")
    return result

# Test cases from https://codesignal.com/learn/course/91/unit/2/practice/3
assert solution(113224) == 2 # From example in the introduction section
assert solution(444) == 2    # From example in the introduction section
assert solution(33333888) == 6
assert solution(13579) == 0
assert solution(345672) == 0
assert solution(22333555) == 5
assert solution(100000) == 4
assert solution(10) == 0
assert solution(98876) == 1
assert solution(4444) == 3
assert solution(1) == 0
```

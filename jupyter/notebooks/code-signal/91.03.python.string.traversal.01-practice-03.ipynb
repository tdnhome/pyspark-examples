***Task***
In this task, you are given a string s, and your goal is to produce a new string following a specific pattern. You are to take characters in sets of three, reverse the characters in each set, and then place them back into the string in their original positions, preserving the reverse order within each set. If 1 or 2 characters remain at the end (because the length of the string is not divisible by 3), they should be left as they are.

The string s contains only lowercase English letters, with its length ranging from 1 to 300, inclusive.

***Example***
- For the input 'abcdef', the output should be 'cbafed'. 
- For the input 'abcdefg', the output should be 'cbafedg'.



```python
def reversed_triple_chars(inputString: str) -> str:
    result = ''
    i = 0
    while i < len(inputString):
        result += extract(inputString, i)
        i += 3
    print(f"Result: {result}")
    return result


def extract(s, p):
    str_len = len(s)
    if p < str_len - 2:
        a, b, c = s[p], s[p+1], s[p+2]
        return c+b+a
    else:
        return s[p:]
        
# Test cases from https://codesignal.com/learn/course/91/unit/3/practice/3
assert reversed_triple_chars("abcdef") == "cbafed"   # From the example in the introduction section
assert reversed_triple_chars("abcdefg") == "cbafedg" # From the example in the introduction section
assert reversed_triple_chars("s") == "s"
assert reversed_triple_chars("reversedtriplechars") == "versretdepircelrahs"
assert reversed_triple_chars("abc") == "cba"
assert reversed_triple_chars("hello") == "lehlo"
assert reversed_triple_chars("abcdefg") == "cbafedg"
assert reversed_triple_chars("hellopython") == "lehpolhtyon"
assert reversed_triple_chars("ab") == "ab"
```

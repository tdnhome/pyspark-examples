## Introduction

***Introduction***

In this task, we are manipulating sentences and strings using nested loops. 

***Task***

- You will be given a string representing a sentence where words are separated by spaces. 
- This sentence string will have a maximum length of 500 characters, including spaces.
- Your objective is to write a Python function that selects the even-indexed characters of words containing an odd number of characters.

Subsequently, these characters must be combined into a single string in the order they appear in the sentence, but the final output string will be reversed end-to-end.

***Example***

- Input sentence: "Coding tasks are fun and required"
- Output string: "tssaefnad"
- reversed string:  "danfeasst".
- Reasoning:
  - The words "tasks", "are", "fun", and "and" are selected since they have an odd number of characters
  - The characters 't', 's', 's', 'a', 'e', 'f', 'n', 'a', 'd' at even indexes are chosen
  - Then the result is reversed in the final string.

***Hints***

  - Python indexing begins at 0, so 't' in "tasks" is considered to be at an even index. 
  - Single-character words must also be taken into consideration for this task.


### Hint: Reverse string using string slicing: string[::-1]

How it works: 
- The slicing format is [start:stop:step]. 
- Leaving start and stop blank tells Python to look at the entire string.
- Providing a step of -1 instructs Python to step through the characters backward

```python
text = "Python"
reversed_text = text[::-1]

print(reversed_text)  # Output: nohtyP
```

### Solution

```python
def solution(sentence):
    result = ""
    words = sentence.split(' ')
    for word in words:
        # confirms whether the length of the word is odd
        if len(word) % 2 == 1:
            # loop over event characters
            for i in range(0, len(word), 2):
                result += word[i]
    
    # return the revered string
    r = result[::-1]
    print(f"result: {r}")
    return r


# Test from the example section
assert solution("Coding tasks are fun and required") == "danfeasst"

# Test from https://codesignal.com/learn/course/93/unit/2/practice/1
assert solution("etrnal") == ""
assert solution("awe-inspiring") == "giisiea"
assert solution("a a") == "aa"
assert solution("python coding") == ""
assert solution("Hello, world!") == ""
assert solution("Mastering Advanced Looping and Implementation") == "dagioLgiesM"
assert solution("Stay hungry, Stay foolish.") == ",rnh"
assert solution("! o G e h C o s i M P") == "PMisoCheGo!"
assert solution("abcdefghijklmnopqrstuvwxyz") == ""
```

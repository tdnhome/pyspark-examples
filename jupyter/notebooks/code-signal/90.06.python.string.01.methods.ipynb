## split() Method

- Constructing strings frequently entails dividing them into smaller sections or 'tokens'.
- The split() function in Python achieves this goal by breaking a string into a list of substrings using a specified delimiter.
- If no delimiter is provided, it splits the string by a single whitespace character.

```python
# Default delimiter is whitespace
sentence = 'Python is fun!'
words = sentence.split() # no delimiter provided, splitting by whitespace
print(words)  # Output is a list of words: ['Python', 'is', 'fun!']
```

```python
# Providing a delimiter
data = 'John,Doe,35,Engineer'
info = data.split(',') # provided a ',' delimiter
print(info)  # Output: ['John', 'Doe', '35', 'Engineer']
```

## join() Method

- Python's join() method concatenates, or 'joins', strings into a single string

```python
# The delimiter is given on the .join method.
words = ['Programming', 'with', 'Python', 'is', 'exciting!']
sentence = ' '.join(words) # Delimiter is a space
print(sentence)  # Output: 'Programming with Python is exciting!'
sentence = '_'.join(words) # Delimiter is a underscore
print(sentence)  # Output: 'Programming_with_Python_is_exciting!'
sentence = '-'.join(words) # Delimiter is a dash
print(sentence)  # Output: 'Programming-with-Python-is-exciting!'
```

## strip(), lstrip(), rstrip Methods

- Discerning extra spaces in strings can prove challenging, and they may lead to problems.
- Python's strip() method removes **leading** and **trailing** spaces, tab or newline characters from a string
- Note that whitespaces withing the string are not removed

***Variations***

- lstrip() remove spaces, tabs, and newline characters from the left of a string
- rstrip() remove spaces, tabs, and newline characters from the right of a string


```python
name = '\t\n    John  Doe    \t\n'
name = name.strip()
print(name)  # Output: 'John  Doe'
```

```python
# lstrip() and rstrip()
name = '    John Doe    '
print(f"'{name.lstrip()}'")  # Output: 'John Doe    '
print(f"'{name.rstrip()}'")# Output: '    John Doe'
```

***Task***
- Let's imagine you are given a string that contains a series of words separated by a hyphen ("-").
- Each word in the string can be a lowercase letter from 'a' to 'z' or a set of digits representing a number from 1 to 26.
- Your task is to parse this string and swap the type of each word:
  - Convert numbers into their corresponding English alphabet letters
  - Convert letters into their numerical equivalents. This means '1' should convert to 'a', and 'a' should convert to '1'.
- You need to return a new string with the converted words, rejoined with hyphens.

***Constraints***
- Ensure you maintain the original order of the words from the input string in your output string.
- The input string's length should range from 1 to 1000 for this exercise.
- The string will never be empty, always containing at least one valid lowercase letter or numerical word.

Remember, the transformation of words should be limited to converting numbers from 1 to 26 into their corresponding letters from 'a' to 'z', and vice versa.

***Example***
- Input: "1-a-3-c-5"
- Expected Output: "a-1-c-3-e".


```python
letter_ord_offset =  ord("a") - 1
def solution(s):
    chars = s.split("-")
    new_chars = []
    
    # Using traditional for loop:
    # for char in chars:
    #    new_chars.append(transform_char(char))

    # Using List Comprehension
    new_chars = [''.join(transform_char(char)) for char in chars] # See 92.01.python.string.operations.00-intro.ipynb

    result = "-".join(new_chars)
    print(f"Result: {result}")
    return result


def transform_char(char):
    if char.isdigit():
        return to_letter(int(char))
    else:
        return to_number(char)


def to_number(char):
    return str(ord(char) - letter_ord_offset)


def to_letter(n):
    return chr(n + letter_ord_offset)



# Test cases from https://codesignal.com/learn/course/92/unit/2/practice/1
assert solution("1-a-3-c-5") == "a-1-c-3-e"
assert solution("1-2-3-4-5") == "a-b-c-d-e"
assert solution("a-b-c") == "1-2-3"
assert solution("1-a-3-c-5") == "a-1-c-3-e"
assert solution("z-y-x-w-v") == "26-25-24-23-22"
assert solution("a-26-b-25-c-24") == "1-z-2-y-3-x"
assert solution("13-9-14-15") == "m-i-n-o"
assert solution("12-1-18-9-1") == "l-a-r-i-a"
assert solution("19-15-12-21-20-9-15-14") == "s-o-l-u-t-i-o-n"
assert solution("a-b-c-1-2-3-x-y-z-24-25-26") == "1-2-3-a-b-c-24-25-26-x-y-z"
assert solution("16-9-20-8-15-14-3-8-1-18-13-1") == "p-i-t-h-o-n-c-h-a-r-m-a"
```

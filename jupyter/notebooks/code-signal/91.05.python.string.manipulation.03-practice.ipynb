***Task***
- Write a Python function that finds repeating two-character patterns in a string.
- The function should identify when the same pair of characters appear next to each other in the string and count how many times each pair repeats consecutively.
- The function should return a new string that lists each pair followed by the number of times it repeats consecutively.

***Example***
- Let's break down the input string "aaabbabbababacab":
- Divide the string into pairs:
  - "aa"
  - "ab"
  - "ba"
  - "bb"
  - "ab"
  - "ab"
  - "ac"
  - "ab"
- Note the consecutive pairs:
  - "ab" appears twice consecutively in the middle.
  - Therefore, the output string will be: "aa1ab1ba1bb1ab2ac1ab1".

Similarly, for the input string "aaababbababaca", the output should be "aa1ab2ba3ca1".

***Key points***
- The input string always has an even number of characters.
- The string contains only lowercase letters.
- The string length can be up to 500 characters.
- Focus on finding consecutive repetitions of the same two-character patterns.

## Solution

```python
def solution(s):
    groups = ""
    current_group_char = None
    current_group_length = 0
    start = 0              # inclusive
    stop = len(s)          # exclusive
    step = 2               # on each loop, read 2 characters

    # Note: input string always has an even number of characters
    for i in range(start, stop, step):
        char = s[i] + s[i+1]
        if char == current_group_char:
            current_group_length += 1
        else:
            if current_group_char is not None:
                groups += current_group_char + str(current_group_length)
            current_group_char = char
            current_group_length = 1
    if current_group_char is not None:
        groups += current_group_char + str(current_group_length)

    print(f"Result: {groups}")
    return groups

# Test cases from https://codesignal.com/learn/course/91/unit/5/practice/3
assert solution("aaabbabbababacab") == "aa1ab1ba1bb1ab2ac1ab1"
assert solution("aaababbababaca") == "aa1ab2ba3ca1"
assert solution("abcabcabcabcab") == "ab1ca1bc1ab1ca1bc1ab1"
assert solution("ab") == "ab1"
assert solution("ccddaaeeff") == "cc1dd1aa1ee1ff1"
assert solution("eeffgg") == "ee1ff1gg1"
```

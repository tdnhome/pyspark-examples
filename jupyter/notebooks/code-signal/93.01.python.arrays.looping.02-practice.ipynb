## Introduction

***Introduction***

You are given two lists: 
- sourceArray: consisting of n tuples, where n is an integer such that 1 ≤ n ≤ 100
- searchArray, consisting of m tuples, where m is an integer such that 1 ≤ m ≤ 500
- Each tuple in both arrays contains two elements: an integer identifier and a string.
- The identifiers in both arrays range from 1 to 100, inclusive.
- The strings in sourceArray consist of alphanumeric characters with lengths ranging from 1 to 100, inclusive.
- The strings in searchArray have lengths ranging from 1 to 500, inclusive.

***Task***

- function: stringSearch(sourceArray, searchArray)
- inputs: two arrays as input
- return: an array that includes all tuples from sourceArray for which
  - its string is a substring of at least one string in any tuple from searchArray and
  - the identifier of the source tuple is less than or equal to the identifier of the search tuple.

The order in which the tuples appear in the result should reflect their original order in the sourceArray. If no matches are found, the function should return an empty array.

***Example***

- SourceArray = [(1, 'abc'), (2, 'def'), (3, 'xyz')]
- SearchArray = [(1, 'abcdef'), (5, 'uvwxy')],
- The function should return [(1, 'abc')]
- Reasoning:
  - since 'abc' and 'def' are substrings found in 'abcdef', but 'def' is associated with 2 in sourceArray which is not less than or equal to 1 in searchArray.
  - The string 'xyz' is not found in either 'abcdef' or 'uvwxy', so it is not included in the result.

***Hint***
This task requires mastery of skills in nested looping and array manipulation, especially in the context of searching for a string within other strings.

### Examples of string.find() and string.index() functions

```python
# string.find(): Returns the lowest starting index if found, or -1 if not found.
# string.index(): Returns the starting index if found, but raises a ValueError if not found.

sourceStr1 = "abc"
sourceStr2 = "def"
sourceStr3 = "xyz"
searchStr = "abcdef"

print(searchStr.find(sourceStr1)) # print 0 
print(searchStr.find(sourceStr2)) # print 3
print(searchStr.find(sourceStr3)) # print -1 (NOT FOUND)

print(searchStr.index(sourceStr1)) # print 0 
print(searchStr.index(sourceStr2)) # print 3
#print(searchStr.index(sourceStr3)) # print ValueError: substring not found
```

### Solution

```python
def stringSearch(sourceArray, searchArray):
    print()
    print("Start stringSearch ....")
    result = [] # Start with an empty list, we will add tuples in the loop below

# Solution Building: Step 2
# --------------------------    
    for sourcePair in sourceArray:
        match_pairs = find_match_pairs(searchArray, sourcePair)
        result.extend(match_pairs)
    print(f"result: {result}")
    return result


# note: match_pairs is defined as a set to avoid duplicate pairs
def find_match_pairs(searchArray, sourcePair):
    print(f"find_match_pairs: searchAray: {searchArray}, sourcePair: {sourcePair}")
    match_pairs = set()
    for searchPair in searchArray:
        searchNum, searchStr = searchPair
        sourceNum, sourceStr = sourcePair
        print(f"searchNum, searchStr: {searchNum}, {searchStr}")
        print(f"sourceNum, sourceStr: {sourceNum}, {sourceStr}")
        if searchStr.find(sourceStr) > -1:
            print(f"searchStr.find(sourceStr) > -1")        
            if sourceNum <= searchNum:
                match_pairs.add(sourcePair)
    return match_pairs
    
# Test from the example in the introduction section
sourceArray = [(1, 'abc'), (2, 'def'), (3, 'xyz')]
searchArray = [(1, 'abcdef'), (5, 'uvwxy')]
expected_output = [(1, 'abc')]
assert stringSearch(sourceArray, searchArray) == expected_output

# Test from https://codesignal.com/learn/course/93/unit/1/practice/2
sourceArray = [(1, 'abc')]
searchArray = [(2, 'def')]
expected_output = []
assert stringSearch(sourceArray, searchArray) == expected_output

sourceArray = [(1, 'abc'), (2, 'def')]
searchArray = [(3, 'abc'), (4, 'def')]
expected_output = [(1, 'abc'), (2, 'def')]
assert stringSearch(sourceArray, searchArray) == expected_output

sourceArray = [(1, 'abc'), (2, 'def'), (3, 'ghi'), (4, 'jkl')]
searchArray = [(5, 'mnopqr')]
expected_output = []
assert stringSearch(sourceArray, searchArray) == expected_output

sourceArray = [(1, 'abc'), (2, 'def'), (3, 'ghi')]
searchArray = [(4, 'abcdefghi'), (5, 'defghiabc')]
expected_output = [(1, 'abc'), (2, 'def'), (3, 'ghi')]
assert stringSearch(sourceArray, searchArray) == expected_output

sourceArray = [(1, 'var'), (2, 'ans'), (3, 'tes')]
searchArray = [(4, 'variant'), (5, 'answertest'), (6, 'tesla')]
expected_output = [(1, 'var'), (2, 'ans'), (3, 'tes')]
assert stringSearch(sourceArray, searchArray) == expected_output

sourceArray = [(1, 'ab'), (2, 'bc'), (3, 'cd')]
searchArray = [(4, 'abcd'), (5, 'bcda')]
expected_output = [(1, 'ab'), (2, 'bc'), (3, 'cd')]
assert stringSearch(sourceArray, searchArray) == expected_output
```

## Task

- Our objective is to create a function that operates on an integer input. 
- Given an integer, n, we need to calculate and return the **sum of its even** digits **without converting n into a string**.
- For instance, if n equals 4625, the output should be 12 because the sum of the even digits 4, 6, and 2 equals 12.
- Keep in mind that n will always be a positive integer between 1 and 10**8

## Steps

Solution Building: Step 1
- To start, we need the basic structure of our function, where we begin by defining a variable digit_sum to keep track of the sum of even digits.


Solution Building: Step 2
- The tool we've chosen to traverse through the digits of the input integer n is the while loop, which is set to run as long as n is greater than zero.

Solution Building: Step 3
- Inside our loop, we'll first extract the last digit of n using the modulo operation (n % 10).
- If the digit is even, we'll increase our digit_sum by this digit.
- After processing a digit, we'll then chop off the last digit of n using integer floor division (n // 10), which allows the while loop to proceed to the next digit.

## Code

```python
def solution(n):
    digit_sum = 0  # Step 1
    while n > 0:   # Step 2
        # Step 3
        digit = n % 10         # Extract the last digit of n
        if digit % 2 == 0:     # Check if the digit is even
            digit_sum += digit # If so, add it to the sum
        n = n // 10            # We are done with the last digit, remove it and continue next loop
    print (f"Result: {digit_sum}")
    return digit_sum

assert solution(4625) == 12
```

## Notes

- The // operator performs floor division (or integer division), returning the largest integer less than or equal to the result.  It discards the fractional part and rounds down toward negative infinity.
- n = n // 10 is the way to remove the last digit from an integer

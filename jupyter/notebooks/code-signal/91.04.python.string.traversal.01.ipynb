***Task***
- You are provided with an array of n integers, where n can range from 1 to 200, inclusive.
- Your task is to create a new array that takes two pairs of 'opposite' elements from the original array at each iteration
- Starting from the center and moving towards both ends, to calculate the resulting ***multiplication*** of each pair.

***Hints***
- By 'opposite' elements, we mean pairs of elements symmetrically located relative to the array's center.
- If the array's length is odd, the center element doesn't have an opposite and should be included in the result array as is.
- Each element in the array can range from -100 to 100, inclusive.

***Example***
- Input: [1, 2, 3, 4, 5]
- Expected [3, 8, 5].
- Reasoning:
  - The center element 3 remains as it is
  - The multiplication of 2 and 4 is 8
  - he multiplication of 1 and 5 is 5.

```python
def solution(numbers):
    result = []
    mid = len(numbers) // 2 # Middle index if odd; right-middle index if even
    if len(numbers) % 2 == 1: # The count of numbers is odd
        result.append(numbers[mid]) # Adding the middle element to the resulting array
        left = mid - 1  # The left to the middle element
        right = mid + 1 # The right to the middle element
    else:
        left = mid - 1 # Left middle element
        right = mid    # Right middle element

    while left >= 0 and right < len(numbers):
        result.append(numbers[left] * numbers[right])
        left -= 1
        right += 1

    print(f"Result: {result}")
    return result

# Test cases from https://codesignal.com/learn/course/91/unit/4/practice/1
assert solution([1, 2, 3, 4, 5]) == [3, 8, 5]
assert solution([1, -1, 1, -1, 1, -1]) == [-1, -1, -1]
assert solution([-100] * 200) == [10000] * 100
assert solution([5, -7, 2, -9, 1, -3, 8]) == [-9, 2, 21, 40]
assert solution([3, 6, 2, 9, -4, -1, 0, 5, 7]) == [-4, -9, 0, 30, 21]
assert solution([9]*200) == [81]*100
assert solution([5]) == [5]
assert solution([0, 0]) == [0]
```

***Task***
- You are provided with a string of alphanumeric characters in which each number, regardless of the number of digits, is always followed by at least one alphabetic character before the next number appears.
- The task requires you to return a transformed version of the string wherein the first alphabetic character following each number is moved to a new position within the string and characters in between are removed.
- Specifically, for each number in the original string, identify the next letter that follows it, and then reposition that character to directly precede the number. All spaces and punctuation marks between the number and the letter are removed.
- The length of the string s ranges from 3 to 10**6 (inclusive), and the string contains at least one number.
- The numbers in the string are all integers and are non-negative.

***Example***
- Input: "I have 2 apples and 5! oranges and 3 grapefruits."
- Output: "I have a2pples and o5ranges and g3rapefruits."
- Explain:
  - The character 'a' following the number 2 is moved to come before the 2,
  - The 'o' succeeding the 5 is placed before the 5, and
  - The 'g' subsequent to the 3 is repositioned to precede the 3. Punctuation marks and spaces in between are removed.

***Example***
- Input: "I have 2apples and 5a oranges and 3 grapefruits."
- Output: "I have a2pples and a5 oranges and g3rapefruits."
- Explain:
  - Notice how the "g" from "grapefruits" moves in front of the "3" because it's the first letter to appear after that number.
  - How might you use a variable to "hold onto" a number while you search for its following letter? 

***Hints***
- The operation should maintain the sequential order of the numbers and the rest of the text.
- The task is not solely about dividing a string into substrings but also about modifying them.
- This will test your expertise in Python string operations and type conversions.

```python
def solution(s):
    result = ""
    num = ""
    for i in range(len(s)):
        char = s[i]
        if char.isdigit():
            num += char
        elif char.isalpha() or char.isspace():
            result += char
            if num.isdigit():
                result += num
                num = ""

    print(f"Result: {result}")
    return result


# Test cases from https://codesignal.com/learn/course/92/unit/2/practice/3
assert solution("I have 2 apples and 5 oranges and 3 grapefruits.") == "I have a2pples and o5ranges and g3rapefruits."
```

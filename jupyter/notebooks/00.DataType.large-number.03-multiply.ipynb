- Write a Python function to multiply two extremely large positive integers.

  - Restriction: you should perform the multiplication without converting the entire strings into integers

***Hints***
- Keep in mind that the elements of the string are digits in the range from 0 to 9, inclusive.
- When multiplying numbers manually, we align the numbers vertically and multiply each digit of the first number with each digit of the second number, starting from the rightmost digits, and add the results after shifting appropriately.



## Solution

### Helper: Floor Division and Modulo Operators

```python
result = 28
print(result // 10) # floor division
print(result % 10)  # modulo - modulus - remainder
```

### Helper: Add n numbers of 0 to end of string

```python
def combine(s, factor):
    n = s + ('0' * factor)
    return n

print(combine("3", 2))
```

### Solution Building

```python
# Solution Building: Step 1
# -----------------------------------
# Before we start coding, let's consider the strategy we're going to adopt. 
# You may recall that each digit in a number has a value, and the position of the digit determines 
# its influence on the total value of the number. This system is called place-value notation.

# The first step requires initialization of our variables. We'll use two indices, i and j, 
# to point to the current digit in num1 and num2, respectively. 
# We'll also need a variable carry to hold carryovers from each addition operation. 
# Lastly, we'll employ a list named res to hold our result, where each digit from the addition is appended at the front.

def solution(num1, num2):
    print()
    print(f"Start solution: {num1, num2} ...")
    carry = 0
    products = [] # the resulting list of integers (as strings) in with the order of significant digits reversed

# Solution Building: Step 2
# -----------------------------------
# Scanning through num1 and num2 from right to left, from the least significant digit to the most significant one.
# For each iteration, we extract the digits n1 from num1 and n2 from num2 and multily them
# The the result (products) are stored on the products array
    num1_loop = len(num1) - 1
    num2_loop = len(num2) - 1
    for i in range(num1_loop, -1, -1):
        carry = 0
        n1 = int(num1[i]) # n1 is a single digit integer
        sub_products = []
        for j in range(num2_loop, -1, -1):
            n2 = int(num2[j]) # n2 is a single digit integer
            # find the product of 2 single digit integers
            sub_product, carry = mul(n1, n2, carry)
            sub_products.append(str(sub_product))   # convert to string so we an do ''.join() later
            
        # Carry is add to the end of the reversed list (the front of the product)
        if carry != 0:
            sub_products.append(str(carry))
            
        # Join the sub products to a single product and add it to the list of products 
        product = ''.join(sub_products[::-1])  # reverse the list and join into a single string
        print(f"Product: {product}, carry: {carry}")
        products.append(product)
        
    # res is the list integers (as strings) in with the order of significant digits reversed
    print(f"Products: {products}")

# Solution Building: Step 3
# -----------------------------------
# Combining products, the product array into a single number
    
    total = 0

    # Special case: only one product
    if (len(products) == 1):
        # Special case: all 0s.  For example: solution("0", "500")
        total = products[0].lstrip('0') or '0'
    else:
        for i in range(len(products)):
            significant = i
            product = products[i] + ('0' * significant)
            total = combine(total, product)
    
    print(f"Result: {total}")
    return str(total)


def mul(n1, n2, carry):
    '''
    Multiply given n1 and n2, both are single digit integers with the given carry applied.
    '''
    product = 0
    new_carry = 0
    r = (n1 * n2) + carry
    if (r >= 10):
        new_carry = r // 10  # Floor division operator
        product = r % 10     # Modulo operator
    else:
        product = r
        new_carry = 0

    print(f"mul: n1: {n1}, n2: {n2}, carry: {carry}, r: {r} product: {product}, new_carry: {new_carry}")
    return (product, new_carry)


def combine(total, r2):
    '''
    Combine (or add) 2 strings representing 2 big integers.
    This function was introduced on the lecture of large-numbers
    '''
    num1 = str(total)
    num2 = r2 
    i = len(num1) - 1
    j = len(num2) - 1
    carry = 0
    res = []

    while i >= 0 or j >= 0 or carry > 0:
        # If both i < 0 and j < 0 and carry > 0, then carry will become the last left digit (most significant)
        n1 = int(num1[i]) if i >= 0 else 0	# set to 0 if there is no number in the current position
        n2 = int(num2[j]) if j >= 0 else 0  # set to 0 if there is no number in the current position
        total = n1 + n2 + carry
        if total > 9:
            carry = 1
        else:
            carry = 0
        curr = total%10    # total%10 will take the remainder for the current digit
        res.append(str(curr))
        i -= 1
        j -= 1        
    return ''.join(res[::-1])  # reverse the list and join into a single string


# Tests 
assert solution("12", "1") == "12"
assert solution("12", "10") == "120"
assert solution("12", "100") == "1200"
assert solution("1", "12") == "12"
assert solution("10", "12") == "120"
assert solution("100", "12") == "1200"
assert solution("999", "12") == "11988"

# Tests from https://codesignal.com/learn/course/93/unit/4/practice/3
assert solution("123456789", "987654321") == "121932631112635269"
assert solution("123", "456") == "56088"
assert solution("100", "200") == "20000"
assert solution("500", "500") == "250000"
assert solution("1000000000", "1000000000") == "1000000000000000000"
assert solution("999999999", "1") == "999999999"
assert solution("0", "500") == "0"
assert solution("1", "1") == "1"
assert solution("9", "9") == "81"
assert solution("111111111", "111111111") == "12345678987654321"
```

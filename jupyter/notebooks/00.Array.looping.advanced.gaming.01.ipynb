## Introduction

***Introduction***

You are given an array of n integers, ranging from 1 to 100 inclusive. 
- **Each integer represents a player's progress** on a linear gameboard, indicating how many steps they can move to the right.
- There exist several **obstacles**, represented by negative integers.

***Task***

Your task is to **transform** the array (the gameboard):
- If an integer can lead the player to an obstacle on its right (within the range of its value), replace the number with the index of the obstacle.
- If the number represents an obstacle (a negative integer), replace it with -1.
- If none of these conditions are met, retain the original integer.

***Hints***
- This task is an innovative take on our previous analysis lesson, implementing a "Move Until Obstacle" game.
- Your array will have no more than 500 elements, and the elements in the array range from -100 to 100, inclusive. 

***Example***

- input = [3, 2, -3, 1, 2]
- output = [2, 2, -1, 1, 2]
- **Reasoning**
  - Replace the first position with 2 because a player at the first position can move 3 steps but will hit the obstacle at the 2nd index.
  - Replace the second position with 2 because a player at the second position can move 2 steps but will hit the obstacle at the 2nd index.
  - Replace the negative number -3 at the third position with -1 because it represents an obstacle.
  - Keep the number 1 at the fourth position as there are no obstacles in its range.
  - Keep the number 2 at the fifth position as there are no further positions or obstacles to impact it.

```python
def solution(numbers):
    result = []
    
    for i in range(len(numbers)):
        number = numbers[i]
        # Replace obstacle (a negative integer) with -1
        if number < 0: 
            number = -1
        else:
            # tep to the right from the current position
            for j in range (i+1, i+number):
                if j < len(numbers):
                    # If an integer can lead the player to or cross an obstacle on its right,
                    # replace the number with the index of the obstacle.
                    print(f"i={i}, number={number}, j={j}")
                    if numbers[j] < 0:
                        number = j
                        break;
        result.append(number)

    print(f"Result: {result}")
    return result


# Test case from example on the introdution section
assert solution([3, 2, -3, 1, 2]) == [2, 2, -1, 1, 2]


```

Comprehensions (list, dict, set, generator): Provide a concise, idiomatic Python way to create new iterables, often preferred over map() and filter() for clarity.

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

References:
1. https://www.w3schools.com/python/python_lists_comprehension.asp

```python
# Without using List Comprehension
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
  if "a" in x:
    newlist.append(x)

print(newlist)
```

```python
# Using List Comprehension
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x if x != "banana"] # with multiple filters = AND
print(newlist)
```

```python
# Using List Comprehension (same as above but use 'and')
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x and x != "banana"] # use and (you can replace with 'or'
print(newlist)
```

```python
# Using List Comprehension (same as above but use 'or'
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x or x == "cherry"] # ORing conditions
print(newlist)
```

With list comprehension you can do all that with only one line of code.
1. Syntax: newlist = [expression for item in iterable if condition == True]
   - expression is the current item in the iteration,
     but it is also the outcome, which you can manipulate before
     it ends up like a list item in the new list
   - expression can also contain conditions to manipulate the outcome
   - iterable can be any iterable object, like a list, tuple, set etc.
   - condition: The condition is like a filter that only accepts the items that evaluate to True.
2. Result: a new list, leaving the old list unchanged.

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

# Example of expression (change all to upper case)
print([x.upper() for x in fruits])

# Example of using range as iterable
print([x for x in range(10) if x < 5])

# Example of expression contains condition that change banana to green banana
print([x if x != "banana" else "green banana" for x in fruits])
print([x if x%2==0 else "None" for x in range(1, 11)])

# Guess what I try to do here
print([num*3 if num % 2 != 0 else "even" for num in range(1,11)])
```

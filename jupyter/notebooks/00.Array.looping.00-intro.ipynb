## Solution

```python
# Solution Building: Step 1
# --------------------------    
# Before we dive into writing code, let's break down the problem. 
# It appears to be an ideal candidate for nested looping.
# Begin by initializing an empty list called result, which will store our pairs.

# It's always prudent to set up your function and data storage first!
def solution(list1, list2):
    result = [] # Start with an empty list, we will add tuples in the loop below

# Solution Building: Step 2
# --------------------------    
# In this step, our focus shifts to the creation of nested loops. 
# We'll need to iterate over both lists simultaneously. 
# To do this efficiently, we'll use nested loops. 
# One outer loop will pick an element from the first list, 
# and an inner loop will skim through each element of the second list.
# In this context, i represents each element in list1, and for each i, j represents each element in list2.
    for i in list1:
        for j in list2:
            
# Solution Building: Step 3
# --------------------------    
# Now that we have our loops established, let's introduce the logic inside. 
# Here's where we make a crucial check: 
# If the item i from list1 < item j from list2, append the pair (i, j) to our result list.
# During each iteration of our inner loop, we make this check and store the pairs that meet our condition.

            if i < j:
                result.append((i, j))

    return result

# Test from the example in the introduction section
list1 = [1, 3, 7]
list2 = [2, 8, 9]
expected = [(1, 2), (1, 8), (1, 9), (3, 8), (3, 9), (7, 8), (7, 9)]
result = solution(list1, list2)
print(f"expected: {expected}")
print(f"result:   {result}")
assert result == expected
```

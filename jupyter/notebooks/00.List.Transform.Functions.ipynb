# 1.1 Python Tranforming iterables

In Python, common functions for transforming iterables include the built-in map() and filter() functions, generator expressions, list comprehensions, and functions from the itertools module. 

In this module, we will look at transforming functions that are not High-Order functions which we have a separate module for (Functional Programing)

## 1.1.1 enumerate()

The enumerate() function is a built-in Python tool used to loop over an iterable (like a list, tuple, or string) while keeping track of the current item's index. It returns an enumerate object that yields pairs in the form of (index, item), which eliminates the need to manually maintain a counter variable.

enumerate() is often used alongside other functional tools such as zip(), sorted(), and reversed() to manipulate sequences. 

Key Features of enumerate()
- Automatic Indexing: It starts counting from 0 by default, but this can be customized.
- Syntax: enumerate(iterable, start=0).
- Return Value: It returns an iterator, specifically an "enumerate object".
- Memory Efficiency: It is efficient for large datasets because it generates the index-value pairs on demand (lazy evaluation).

Advantages
- Readability: It makes code cleaner and more Pythonic compared to using range(len()).
- Safety: It reduces the chance of off-by-one errors in manual counter management.
- Flexibility: It works with any iterable (lists, tuples, strings, files, etc.)

Common Pitfalls
- Tuple Unpacking: enumerate() returns a tuple (index, item). Forgetting to unpack this (e.g., for item in enumerate(data)) will result in the variable holding the tuple instead of just the item.


```python
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits, start=1):  # set start index, default = 0 
    print(index, fruit)
```

```python
# Convert a list to tuple
my_list = ['a', 'b', 'c']
print(list(enumerate(my_list)))
```

```python
# enumerate a dictionary.items
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
for index, (key, value) in enumerate(d.items()):
    print(f"Index: {index}, Key: {key}, Value: {value}")
    
```

```python
# enumerate a dictionary.keys
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
for index, key in enumerate(d.keys()):
    print(f"Index: {index}, Key: {key}")
```

```python
# enumerate a dictionary.values
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
for index, value in enumerate(d.values()):
    print(f"Index: {index}, value: {value}")
```

```python
# enumerate a tuple 
t = (1, 2, 3, 4)
for index, value in enumerate(t):
    print(f"Index: {index}, value: {value}")
```

## 1.1.2 max()

The Python max() function is a built-in tool that returns the largest item in an iterable (like a list or tuple) or the largest of two or more arguments. It supports custom sorting via a key argument, handles lexicographical string comparison, and allows a default value if the iterable is empty. 

Key Features and Usage:
- Basic Usage: max(3, 5, 1, 9) returns 
- Iterables: max([1, 2, 3]) returns 
- String Comparison: max("apple", "banana") returns "banana" (lexicographical).
- Key Argument: Use key=len to find the longest string, or key=str.lower for case-insensitive comparison.
- Default Value: max([], default=0) returns () instead of raising a ValueError. 

Common Use Cases:
- Finding Max in List: max(numbers) is efficient for finding the highest number in a list.
- Dictionary Lookup: max(dict, key=dict.get) returns the key with the highest value. 


```python
print(max(3, 1, 4))
print(max([1, 2, 0]))
print(max(['cake', 'banana']))           # default order for string is lexicographical
print(max(['banana', 'cake'], key=len))  # use len function
```

## 1.1.5 min()

The built-in min() function in Python returns the smallest item from an iterable (like a list or tuple) or the smallest of two or more arguments. It supports customized comparisons using a key function and offers a default value if the iterable is empty. It works on numbers, strings, and other comparable types. 

Key Aspects of min() Function:
- Syntax:
1. min(iterable, *[, key, default]): Finds the minimum in an iterable.
2. min(arg1, arg2, *args[, key]): Finds the minimum among multiple arguments.
- Key Parameter: Allows using a function (e.g., lambda or str.lower) to modify how items are compared (e.g., min(data, key=str.lower)).
- Default Parameter: Returns a specified value if the iterable is empty, preventing a ValueError.
- Performance: It efficiently finds the minimum value in a single pass (time complexity).

Usage Tips:
- Use key=... to find the minimum object based on a specific attribute, such as comparing tuples based on their second element.
- Be cautious when comparing different types, as it will raise a TypeError.
- For finding the minimum value in a SQL database, the MIN() aggregate function is used.


```python
print(min(3, 1, 4))
print(min([1, 2, 0]))
print(min(['cake', 'banana']))           # default order for string is lexicographical
print(min(['banana', 'cake'], key=len))  # use len function
```

## 1.1.2 reversed()

In Python, the built-in functional tool for reversing iterables is the reversed() function. It provides an efficient way to iterate over a sequence in reverse order without modifying the original data. 

Here is a breakdown of the built-in tools for reversing in Python:
1. reversed(seq) Function: Returns a reverse iterator that yields elements in reverse order.

Key Characteristics:
- Memory Efficiency: It is lazy and does not create a full copy of the list, making it in space complexity.
- Versatility: Works on any sequence (list, tuple, string) or any object that supports the __reversed__ method.
- Usage: Often used in for loops. To get a reversed list, you must wrap it in list().

2. list.reverse() Method: Reverses the elements of a list in-place.
Key Characteristics:

- Side Effects: Modifies the original list directly.
- Performance: time complexity, but efficient because it does not create a new list.
- Return Value: Returns None.


```python
# Example of using reversed function.  Note: the orginal list stays the same
original = [1, 2, 3]
for item in reversed(original):
    print(item)
print(f"original: {original}")
```

```python
# Example of using list reverse method.  Note: the original list is reversed
original = [1, 2, 3]
original.reverse()
print(f"original: {original}")
```

## 1.1.3 sorted()

Python provides two primary built-in tools for sorting data: the sorted() function and the list.sort() method. Both are highly optimized, implemented in C, and use the Timsort algorithm, which is a hybrid sorting algorithm derived from merge sort and insertion sort.

1. sorted(): The built-in sorted() function is a versatile, built-in function that returns a new sorted list from any iterable (lists, tuples, dictionaries, strings). 

- Original Data: Remains unchanged.
- Syntax: sorted(iterable, key=None, reverse=False).
- Default: Sorts in ascending order. 

2. list.sort(): The built-in method specifically for lists that modifies the list in-place. 
- Original Data: Modified directly.
- Return Value: Returns None.
- Efficiency: Slightly faster for very large lists as it avoids creating a new list. 

```python
# Examples of using sorted function

# Sorting a list
numbers = [5, 2, 3, 1, 4]
sorted_numbers = sorted(numbers)
print(f"sorted_numbers: {sorted_numbers}")

# Sorting in descending order
desc_numbers = sorted(numbers, reverse=True)
print(f"descending sorted_numbers: {desc_numbers}")

# Sorting with a key (e.g., by string length)
words = ["banana", "pie", "apple"]
sorted_words = sorted(words, key=len)  # ['pie', 'apple', 'banana']
print(f"sorted_words by string length: {sorted_words}")
```

```python
# Examples of using list.sort method

items = [3, 1, 2]
items.sort() 
print(f"sorted items: {items}")
```

## 1.1.4 zip()

The zip() function in Python is a built-in functional tool that pairs elements from two or more iterables (lists, tuples, strings) element-wise, creating an iterator of tuples. It stops at the shortest input iterable. Commonly used for parallel iteration or creating dictionaries, it returns a lazy zip object. 

Key Features of zip()
- Element-wise Pairing: Pairs the 1st elements, then 2nd, etc..
- Iterator Output: Returns a memory-efficient iterator, not a list. Use list(zip()) to view all pairs at once.
- Shortest Input: Stops when the shortest iterable is exhausted.
- Multiple Arguments: Can accept multiple iterables to form n-tuples.
- Unzipping: Use zip(*zipped_list) to reverse the process.

When to Use zip()
- Iterating over multiple lists: Ideal for looping through related lists in parallel without using index counters.
- Data Aggregation: Combining data from different sources (e.g., merging a list of keys and a list of values).

```python
# Basic pairing
names = ["Alice", "Bob"]
scores = [88, 94]
paired = list(zip(names, scores))  # zip to pair the list
print(f"Paring - create list of tuples from given 2 list:")
print(f"  {paired}")
```

```python
# Parallel iteration
print(f"Parallel iteration:")
for name, score in zip(names, scores):  # zip to extract items from multiple lists
    print(f"  {name}: {score}")
```

```python
# Creating a dictionary
print(f"Create dictionary from lists:")
student_scores = dict(zip(names, scores)) # zip to extract items from multiple lists then create a dictionary
print(f"  {student_scores}")
```

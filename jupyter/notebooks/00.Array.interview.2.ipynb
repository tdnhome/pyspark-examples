## Problem Statement

***Introduction***

As an aspiring musician, you are given a chance to create a unique encrypted song. The song's lyrics will be encoded based on specific rules applied to an array of integers and a string.

- You are provided with an array of n integers, where n is between 1 to 500, inclusive. Each integer in the array ranges from −100 to 100, inclusive.
- Accompanying this, you are presented with a string comprising n lowercase alphabetical characters.

***Tasks***

- For the array, calculate the absolute value of each number and multiply it by 2. Accumulate these results until the total exceeds 100.
- For the string, your task is to replace each letter with the preceding alphabetical character and then concatenate these characters in the order they were processed. 

When the accumulated total exceeds 100 or if the next character is a vowel, halt the process and return the transformed string and the remainder of the array in their original order.

The final output should be a tuple with
- the transformed string as the first element and
- the remaining, unprocessed part of the array as the second element

***Hint***

In addition to checking the 2 condition for stoping the loop, you need check the loop index<len(input string) to avoid index-out-of-range issue.

***Example 1***

strings = "bcd" and numbers = [10, 20, 30]:

- For 'b', we get 'a' and the sum becomes abs(10) * 2 = 20.
- For 'c', we get 'b' and the sum becomes 20 + (abs(20) * 2) = 60.
- For 'd', we get 'c' and the sum becomes 60 + (abs(30) * 2) = 120.

Since 120 exceeds 100, we stop and return ('abc', []). If a vowel like 'e' appeared, we would stop even sooner! 🎵

***Example 2***

strings = "bca" and numbers = [10, 20, 30]:

- For 'b', we get 'a' and the sum is 20.
- For 'c', we get 'b' and the sum is 60.
- The next character is 'a', which is a vowel! we don't process 'a' or its corresponding number. The result would be ('ab', [30]). 🎶


## Solution

```python
def solution(strings, numbers):
    SUM_LIMIT = 101    
    sum = 0
    string = ""
    i = 0
    while (i < len(strings)) and (not is_vowel(strings[i])) and (sum < SUM_LIMIT):
        s = chr(ord(strings[i]) - 1)
        sum += abs(numbers[i] * 2)
        string += s
        i += 1

    # Returning a tuple
    #  - string_so_far
    #  - numbers[i:]  -> sequence slicing from i to the end of the sequence
    return string, numbers[i:]

def is_vowel(c):
    if c.lower() in 'aeiou':
        return True
    return False
    
# Test
assert solution('bcd', [10, 20, 30]) == ('abc', [])
assert solution('bca', [10, 20, 30]) == ('ab', [30])
assert solution('abcdef', [10, 20, 30, 40, 50, 60]) == ('', [10, 20, 30, 40, 50, 60])
assert solution('xyzabc', [10, 20, 30, 40, 50, 60]) == ('wxy', [40, 50, 60])
assert solution('bcdefg', [30, 40, 35]) == ('ab', [35])
assert solution('xyzxyz', [10, 20, 30, 40, 50, 60]) == ('wxy', [40, 50, 60])
assert solution('abc', [200]) == ('', [200])
assert solution('z', [50]) == ('y', [])
```

## Problem Statement

***Introduction***

Both camelCase and snake_case are popular naming conventions used in programming to combine multiple words into a single variable or function name. The core difference is how they separate words: camelCase uses capital letters on the second and subsequent words, while snake_case uses underscores.

***Requirement***

- You are given a string containing words.  
- Convert all snake case words to camel case.

***Task***

Write a Python function, toCamelCase(string), that accepts as input string and convert all snake case words to camel case

***Hint***


## Solution

### Example

````python
text = "Learn Python text processing"
words = text.split() 
print(words)

words.append("is")  # note that list.append() return None, so we couldn't pipe append().append() like java
words.append("fun")
sentence = " ".join(words)
print(sentence)  
````

### Task

````python
def toCamelCase(doc):
    words = doc.split()
    updated_words = []
    for word in words:
        updated_word = transform(word)
        updated_words.append(updated_word)
    #print(f"updated_words: {updated_words}")
    return " ".join(updated_words)


def transform(word):
    #print(f"word: {word}")
    prev_letter = ""
    x_word = ""
    for letter in word:
        #print(f"letter: {letter}, prev_letter: {prev_letter}")
        x = letter
        if letter == "_":
            x = ""
        elif prev_letter == "_":
            x = letter.upper()
        x_word += x
        prev_letter = letter
    #print(f"x_word: {x_word}")
    return "".join(x_word)


# Test
print(toCamelCase("This_is a test_Case in myFunction"))
````

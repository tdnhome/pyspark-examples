# Polymorphism in Python

Polymorphism means "many forms" and allows different object classes to share the same method name but execute their own unique behavior. It makes your code highly flexible and reusable by letting you treat different objects uniformly.

There are two primary ways to implement polymorphism in Python Object-Oriented Programming (OOP):
1. Polymorphism with Inheritance (Method Overriding)
2. Polymorphism with Duck Typing

## Polymorphism with Inheritance (Method Overriding)

This traditional approach uses a shared parent class. Subclasses inherit from the parent but override its specific methods to provide unique actions.

```python
# Parent class (an interface)
class PaymentProcessor:
    def process_payment(self, amount):
        # A generic method meant to be overridden
        raise NotImplementedError("Subclass must implement this method")

# Subclass 1
class CreditCardPayment(PaymentProcessor):
    def process_payment(self, amount):
        return f"Processing credit card payment of ${amount} via Bank API."

# Subclass 2
class PayPalPayment(PaymentProcessor):
    def process_payment(self, amount):
        return f"Redirecting to PayPal gateway to pay ${amount}."

# Unified function that accepts any PaymentProcessor object
def checkout(payment_object, total_amount):
    # This single line handles completely different behaviors automatically
    print(payment_object.process_payment(total_amount))

# Usage
card = CreditCardPayment()
paypal = PayPalPayment()

# Demo of using polimorphism
checkout(card, 150)    # Output: Processing credit card payment of $150 via Bank API.
checkout(paypal, 45)  # Output: Redirecting to PayPal gateway to pay $45.
```

## Polymorphism with Duck Typing

Python is a dynamically-typed language and does not require explicit class hierarchies to achieve polymorphism. Under Duck Typing, Python focuses **solely on whether an object has the requested method**, not its exact type. "If it walks like a duck and quacks like a duck, it's a duck."

```python
# Independent class 1 (No shared parent class)
class Dog:
    def speak(self):
        return "Woof!"

# Independent class 2 (No shared parent class)
class Cat:
    def speak(self):
        return "Meow!"

# A loop that treats completely separate classes uniformly
animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
    
```

## Introduction

***Introduction***

- You are given a string (sentence) of n words, with n ranging from 1 to 100, inclusive.
- The words are separated by a single space in the string.
- Each word's character count ranges from 1 to 500, inclusive.
- The string contains lowercase and uppercase alphanumeric characters, spaces, and punctuation.

***Task***

- Your task is to return the most frequently occurring character in each word that has an odd number of characters.
- The resulting characters should be concatenated into a string with their occurrences in the sentence.
- If there are no words with an odd number of characters in the input string, your function should return an empty string.

***Additional Requirements***

- The function should be case insensitive. The lowercase and uppercase characters should be counted as the same character.
- The output should only contain lowercase characters. For example: "Hhi" should return "h" because "h" appears twice in the string even though one is uppercase and one is lowercase.
- In case of a tie in character frequency, return the character that appears first in the word. In the example above, we took 'w' from the word 'world'.

***Example***

- Input : "Hello world this is a demo string"
- Output: "lwa".
- In this string, 'Hello', 'world', and 'a' have an odd number of characters.
- The most frequently occurring character in these words are 'l', 'w', and 'a' respectively.
- When concatenated, they form "lwa".

## Hint: Find the most frequenly occurring char using dictionary and max(dictionary, key=dict.get)

Find the most frequenly occurring char using dictionary and max(dictionary, key=dict.get)

***How it works***
- https://www.geeksforgeeks.org/python/python-get-key-with-maximum-value-in-dictionary/
- https://www.geeksforgeeks.org/python/python-max-function/

***Syntax:*** max(arg1, arg2, *args[, key]) 
- arg1, arg2 : objects of the same datatype
- *args : multiple objects
- key : **function** where comparison of iterable is performed based on its return value

```python
def find_most_frequent_char(word):
    # 1. Initialize the frequency dictionary
    frequency = {}
    
    # 2. Build frequency dictionary
    for c in word:
        char = c.lower()
        frequency[char] = frequency.get(char, 0) + 1

    # 3. Extract character with maximum value 
    return max(frequency, key=frequency.get)

print(find_most_frequent_char("helLo"))
print(find_most_frequent_char("hi"))
```

## Solution

```python
def solution(sentence):
    result = ""
    words = sentence.split(' ')
    for word in words:
        # confirms whether the length of the word is odd
        if len(word) % 2 == 1:
            c = find_most_frequent_char(word)
            result += c
    
    print(f"result: {result}")
    return result


def find_most_frequent_char(word):
    frequency = {}
    for c in word:
        char = c.lower()
        frequency[char] = frequency.get(char, 0) + 1
    return max(frequency, key=frequency.get)


# Test from the example section
assert solution("Hello world this is a demo string") == "lwa"

# Test from https://codesignal.com/learn/course/93/unit/2/practice/2
assert solution("Python is a high-level programming language") == "ar"
assert solution("Mastering Advanced Looping and Implementation") == "moa"
assert solution("nested loops") == "o"
assert solution("Python provides us with a built-in") == "a"
assert solution("Bravo! You've just successfully navigated through") == "ah"
assert solution("Now, use this knowledge as a foundation in your exploration of nested loops") == "ueaoo"
assert solution("Isn't it fascinating") == "ia"
assert solution("Python") == ""
assert solution("a") == "a"
assert solution("high-level") == ""
```
